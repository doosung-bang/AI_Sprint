## 한글이 깨지는 이유

matplotlib는 글자를 그릴 때 "폰트 파일"을 찾아서 그 모양대로 그립니다.

Colab의 기본 폰트에는 한글 글자 모양이 들어 있지 않습니다.
그래서 "월"이라는 글자를 그리려고 폰트를 뒤졌는데 없으니까,
"모양을 못 찾았다"는 표시로 네모를 그립니다.

즉,
- 코드 문제 아님
- 인코딩 문제 아님
- 한글 폰트 파일이 없는 문제

해결 순서
1. 한글 폰트를 설치한다
2. matplotlib에게 그 폰트 파일의 위치를 알려준다
3. 앞으로 이 폰트를 쓰라고 지정한다

## plt.plot 방식의 한계

지금까지 쓴 방식은 이렇습니다.

    plt.plot(...)
    plt.title(...)
    plt.show()

편합니다. 그런데 문제가 있습니다.

`plt.title()` 이라고 쓸 때, 이 제목은 "어느 그림"에 붙는 걸까요?
그림이 하나일 때는 답이 뻔합니다. 그 하나에 붙습니다.

그림이 네 개일 때는요?

plt 방식은 "지금 활성화된 그림"에 붙임
그 "지금"이 뭔지는 코드만 봐서는 알 수 없습니다.

그래서 실무에서는 다른 방식을 씁니다.
그림 객체를 직접 손에 쥐고, 그 객체한테 명령합니다.

```python
fig1 = plt.figure()
plt.plot([1,2,3])

fig2 = plt.figure()
plt.plot([4,5,6])

plt.title("첫 번째 그래프")
```

### C15. Figure와 Axes

```python
# fig  = 도화지 (그림 전체, 액자)
# ax   = 그 위의 그래프 하나 (실제로 선이 그려지는 영역)
fig, ax = plt.subplots()

# 이제 plt 가 아니라 ax 에게 직접 명령합니다.
ax.plot(요일, 매출)

plt.show()

# 각각이 무엇인지 확인해 봄
print("fig 타입:", type(fig))
print("ax 타입:", type(ax))
```

### C16. 제목과 축 이름 붙이기

```python
fig, ax = plt.subplots()

# 선을 그린다. label 은 나중에 범례에 표시될 이름임
ax.plot(요일, 매출, label="매출")

# ax 에게 명령: 제목 붙여라
ax.set_title("커피 강남점 요일별 매출")

# ax 에게 명령: 가로축 이름 붙여라
ax.set_xlabel("요일")

# ax 에게 명령: 세로축 이름 붙여라
ax.set_ylabel("매출(만원)")

# ax 에게 명령: 범례 띄워라 (label 을 준 선들이 목록으로 나옵니다)
ax.legend()

plt.show()
```

### C18. 데이터 준비

```python
import pandas as pd
import numpy as np

# --- 데이터 1: 강남점 7일치 (요일별 요약) ---
week = pd.DataFrame({
    "요일": ["월", "화", "수", "목", "금", "토", "일"],
    "매출": [120, 135, 128, 150, 210, 320, 290],     # 단위: 만원
    "방문객": [80, 88, 84, 96, 130, 190, 175],        # 단위: 명
})

# --- 데이터 2: 메뉴별 판매량 ---
menu = pd.DataFrame({
    "메뉴": ["아메리카노", "라떼", "콜드브루", "마카롱", "샌드위치"],
    "판매량": [420, 260, 180, 95, 130],
})

print(week)
print()
print(menu)
```

## B-1. 시간에 따라 어떻게 변했나 - 꺾은선

### 질문

### 한국어 설계

2. 가로축은 요일, 세로축은 매출로 꺾은선을 그린다
3. 각 점이 어디인지 보이게 마커를 찍는다
4. 제목과 축 이름을 붙인다
5. 화면에 띄운다

### 예상되는 그림
- 월~목은 낮고 평평
- 금요일부터 상승
- 토요일 최고점, 일요일 소폭 하락

### C20. B-1 꺾은선

```python
# 1. 도화지와 그래프 칸을 만든다 (figsize 로 가로세로 크기 지정, 단위는 인치)
fig, ax = plt.subplots(figsize=(7, 4))

# 2~3. 꺾은선 + 마커
#      marker="o" : 각 데이터 점에 동그라미 표시
ax.plot(week["요일"], week["매출"], marker="o", label="매출")

# 4. 제목과 축 이름
ax.set_title("커피 강남점 요일별 매출")
ax.set_xlabel("요일")
ax.set_ylabel("매출(만원)")
ax.legend()

# 5. 띄운다
plt.show()
```

## B-2. 항목끼리 누가 더 큰가 - 막대

### 질문
어떤 메뉴가 가장 많이 팔렸는가?

### 한국어 설계
1. 도화지와 그래프 칸을 만든다
2. 가로축은 메뉴 이름, 세로축은 판매량으로 막대를 그린다
3. 제목과 축 이름을 붙인다
4. 화면에 띄운다

### 예상되는 그림
- 아메리카노가 가장 높은 막대 (420)
- 마카롱이 가장 낮은 막대 (95)
- 막대 5개

### C22. B-2 막대

```python
fig, ax = plt.subplots(figsize=(7, 4))

# ax.bar(가로축 항목, 세로축 값)
ax.bar(menu["메뉴"], menu["판매량"], color="tab:orange")

ax.set_title("메뉴별 판매량")
ax.set_xlabel("메뉴")
ax.set_ylabel("판매량(잔/개)")

plt.show()
```

### C23. B-2 보강: 정렬과 가로 막대

```python
# 판매량 기준 내림차순 정렬 (큰 것부터)
menu_sorted = menu.sort_values("판매량", ascending=True)

fig, ax = plt.subplots(figsize=(7, 4))

# barh 는 가로 막대임 항목 이름이 길 때 유리합니다.
ax.barh(menu_sorted["메뉴"], menu_sorted["판매량"], color="tab:orange")

ax.set_title("메뉴별 판매량 (내림차순)")
ax.set_xlabel("판매량(잔/개)")

# 가로 막대는 위아래가 뒤집혀 보이므로, 1등을 맨 위로 올립니다.
ax.invert_yaxis()

plt.show()
```

### C26. B-3 보강: 점이 겹칠 때

```python
# 점이 많으면 겹쳐서 몇 개인지 안 보임 alpha 로 투명도를 줍니다.
# alpha=1 이 완전 불투명, 0 이 완전 투명임

# 실습용으로 점을 많이 만들어 봄
rng = np.random.default_rng(0)
x = rng.normal(100, 15, 300)
y = x * 2.4 + rng.normal(0, 20, 300)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].scatter(x, y)                    # 투명도 없음
axes[0].set_title("alpha 없음")

axes[1].scatter(x, y, alpha=0.3, s=20)   # 투명도 + 점 크기 축소
axes[1].set_title("alpha=0.3, s=20")

plt.show()
```

### C28. B-4 데이터 준비

```python
# 3개 지점 x 28일 = 84일치 데이터를 만듭니다.
rng = np.random.default_rng(0)

# np.repeat: 각 값을 28번씩 반복 -> 강남 28개, 홍대 28개, 판교 28개
지점 = np.repeat(["강남", "홍대", "판교"], 28)

# np.tile: 리스트 전체를 12번 반복 -> 월화수목금토일 월화수목금토일 ...
요일 = np.tile(["월", "화", "수", "목", "금", "토", "일"], 12)

# 지점별 평일 기준 평균 방문객
기본방문 = np.repeat([120, 90, 70], 28)

# 토, 일이면 1.6배, 아니면 1배
주말배수 = np.where(np.isin(요일, ["토", "일"]), 1.6, 1.0)

# 평균 근처에서 흔들리는 값을 만듭니다 (normal = 정규분포 난수)
방문객 = (기본방문 * 주말배수 + rng.normal(0, 10, 84)).round()

sales = pd.DataFrame({"지점": 지점, "요일": 요일, "방문객": 방문객})

# 객단가(손님 1명당 매출). 주말에는 회전이 빨라져 객단가가 낮아진다고 가정합니다.
객단가 = np.where(np.isin(요일, ["토", "일"]), 2.0, 2.4)

# 매출 = 방문객 x 객단가 + 약간의 흔들림
sales["일매출"] = (sales["방문객"] * 객단가 + rng.normal(0, 20, 84)).round()

print(sales.shape)
print(sales)
```

### C29. B-4 히스토그램

## Part C 정리

| 하고 싶은 것 | 코드 |
|---|---|
| 칸 1개 | `fig, ax = plt.subplots()` |
| 가로 2칸 | `fig, axes = plt.subplots(1, 2)` |
| 2행 2열 | `fig, axes = plt.subplots(2, 2)` |
| 왼쪽 칸 지목 | `axes[0]` 또는 `axes[0, 0]` |
| 칸 제목 | `ax.set_title(...)` |
| 전체 제목 | `fig.suptitle(...)` |
| 축 맞추기 | `sharey=True` |
| 간격 정리 | `fig.tight_layout()` |

- 단위가 다르다 -> 칸을 나눈다
- 칸을 나눴다 -> 축을 맞춘다
