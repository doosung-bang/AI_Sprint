### 문제 07. reshape로 같은 값의 모양만 바꾸기

`np.arange(12)`로 값 12개를 만든 뒤, 다음 세 가지 모양으로 각각 바꾸세요.

1. 3행 4열
2. 4행 3열
3. 2행 6열

각 결과와 `shape`, `size`를 출력하고, **원본 배열이 바뀌지 않았는지**도 확인하세요.

### 문제 08. reshape가 실패하는 이유를 직접 확인하고 고치기

값 12개를 `reshape(5, 3)`으로 바꾸려고 하면 어떻게 될까요?

**먼저 예측하세요.** 성공할까요, 실패할까요? 실패한다면 왜일까요?

그다음 `try / except ValueError`로 감싸서 실행하고, 오류의 **종류 이름**과 **이유**를 출력하세요.
마지막으로 값 개수를 바꾸지 않고 성공하는 모양 하나를 골라 다시 실행하세요.

**[한글 설계] — 코드를 쓰기 전에 여기부터 채우세요**

```text
1.
2.
3.
```

````python
import numpy as np

base = np.arange(12)


# ===== 여기에 코드를 작성하세요 =====
try:
    five_by_three = base.reshape(5, 3)

except ValueError as error:
    print("예상한 오류 종류:", type(error).__name__)
    print("이유: 값 12개를 15칸 모양에 넣을 수 없습니다.")

else:
    print("5행 3열:\n", five_by_three )

retry = base.reshape(6, 2)
print(retry)
print("retry shape:", retry.shape, "size:", retry.size)
````

### 문제 23. argsort로 상위 3명 뽑기

바리스타 이름과 제조 잔 수가 짝을 이루고 있습니다.

```python
baristas = np.array(["민준", "서연", "도윤", "하윤", "지호"])
made = np.array([182, 240, 165, 258, 203])
```

제조 잔 수가 **많은 순서로 상위 3명의 이름과 잔 수**를 출력하세요.

`np.sort`로는 이름을 함께 옮길 수 없습니다. **위치를 먼저 정렬**해야 합니다.

**[한글 설계] — 코드를 쓰기 전에 여기부터 채우세요**

```text
1.
2.
3.
```

````python
import numpy as np

baristas = np.array(["민준", "서연", "도윤", "하윤", "지호"])
made = np.array([182, 240, 165, 258, 203])


# [옮기기] argsort는 '정렬된 값'이 아니라 '정렬 순서대로의 위치'를 준다
order_asc = np.argsort(made) #-- 작은값에서 큰값의 순서로 위치 인덱스 번호 반환
print("오름차순 위치:", order_asc)
print("확인 - 그 위치의 값:", made[order_asc])

# [옮기기] 뒤집어서 내림차순 위치로 만들고, 앞 3개만 자른다
top3 = np.argsort(made)[::-1][:3]

print("상위 3명 위치:", top3)
print("상위 3명 이름:", baristas[top3])
print("상위 3명 잔 수:", made[top3])
````

````python
import numpy as np

sold = np.array([
    [42,  88, 61, 25],
    [30,  75, 94, 48],
    [55, 102, 47, 12],
])

# ===== 여기에 코드를 작성하세요 =====
print(f'sold표 \n{sold}')
print(f'sold shape  : {sold.shape}')
time_bonus = np.array([1, 2, 3, 4])
print(f'time_bonus 표 {time_bonus}')
print(f'time_bonus shape : {time_bonus.shape}')
try:
    apply_time_bonus = sold + time_bonus 
    #시간대별 보너스: (3,4)와 (4,) -> 마지막 축이 4로 일치하므로 성공

except ValueError as error:
    print("예상한 apply_time_bonus오류 종류:", type(error).__name__)
    print("이유: shape 형태.")
else:
    print(f' apply_time_bonus: \n{apply_time_bonus} ') 

branch_bonus = np.array([10, 20, 30])
print(f'branch_bonus \n{branch_bonus}')
print(f'branch_bonus.shape : {branch_bonus.shape}')
try:
    applied_branch_bonus = sold + branch_bonus 
    #지점별 보너스: (3,4)와 (3,) -> 마지막 축이 4 vs 3 -> 실패

except ValueError as error:
    print("예상한 branch_bonus 오류 종류:", type(error).__name__)
    

column_bonus = branch_bonus.reshape(3, 1)
print(f'branch_bonus.reshape \n{column_bonus} ')
print(f'branch_bonus.reshape : {column_bonus.shape}')
print("지점 보너스 적용:\n", sold + column_bonus)

````
