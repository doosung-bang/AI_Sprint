## Part 3. 배열의 모양 바꾸기

`reshape`는 값의 순서를 유지하면서 모양을 바꿉니다.

반복 질문: **숫자가 총 몇 개인가?**

10개는 `(2, 5)`, `(5, 2)`, `(1, 10)`으로 바꿀 수 있지만 `(4, 3)`에는 들어가지 않습니다.

````python
base = np.arange(10)
two_by_five = base.reshape(2, 5)
five_by_two = base.reshape(5, 2)

print("원본:", base, "shape=", base.shape)
print("2행 5열:\n", two_by_five)
print("5행 2열:\n", five_by_two)
print("size:", base.size, two_by_five.size, five_by_two.size)
````

### axis 앵커

2차원 표에서:

- `axis=0`: 행 방향으로 내려가며 계산 → **열마다 결과 하나**
- `axis=1`: 열 방향으로 옆으로 가며 계산 → **행마다 결과 하나**

암기 대신 결과 개수로 확인합니다.

````python
axis_demo = np.array([
    [8, 12],
    [7, 1],
    [9, 5],
])

print("원본:\n", axis_demo)
print("axis=0 정렬(열마다):\n", np.sort(axis_demo, axis=0))
print("axis=1 정렬(행마다):\n", np.sort(axis_demo, axis=1))
print("axis=0 합계(열마다):", axis_demo.sum(axis=0))
print("axis=1 합계(행마다):", axis_demo.sum(axis=1))
````

````python
names = np.array(["John", "Mike", "Sarah", "Kate", "Samuel"])
scores = np.array([78, 95, 84, 98, 88])

order = np.argsort(scores)

print("오름차순 위치:", order)
print("오름차순 점수:", scores[order])
print("오름차순 이름:", names[order])
````

## Preview. AI로 이어지는 shape의 역할

이 구간은 맛보기입니다.

목표는 공식을 외우는 것이 아니라 **계산 전에 shape를 확인하는 이유**를 보는 것입니다.

````python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
])
bonus = np.array([1, 2, 3])

print("matrix shape:", matrix.shape)
print("bonus shape :", bonus.shape)
print("결과:\n", matrix + bonus)
````

````python
wrong_bonus = np.array([1, 2])

try:
    matrix + wrong_bonus
except ValueError as error:
    print("예상한 오류 종류:", type(error).__name__)
    print("matrix shape:", matrix.shape)
    print("wrong_bonus shape:", wrong_bonus.shape)
````

````python
fixed_bonus = np.array([1, 2, 3])
fixed_result = matrix + fixed_bonus

print("fixed shape:", fixed_bonus.shape)
print(fixed_result)
````

### 벡터 내적 한 문장

같은 위치끼리 곱한 뒤, 그 결과를 모두 더합니다.

```text
[1, 2, 3] · [4, 5, 6]
= 1×4 + 2×5 + 3×6
= 32
```

`*`는 원소별 곱셈 결과 여러 개, `np.dot()`은 그 곱들을 더한 결과 하나를 냅니다.

````python
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])

elementwise = vector_a * vector_b
dot_value = np.dot(vector_a, vector_b)

print("원소별 곱:", elementwise)
print("곱의 합:", elementwise.sum())
print("내적:", dot_value)
````

````python
A = np.array([
    [1, 2, 3],
    [4, 5, 6],
])
B = np.array([
    [7, 8],
    [9, 10],
    [11, 12],
])

result = np.dot(A, B)

print("A shape:", A.shape)
print("B shape:", B.shape)
print("결과 shape:", result.shape)
print("행렬 내적:\n", result)
print("첫 칸 확인:", 1*7 + 2*9 + 3*11)
````

## Final Challenge — 답안을 보기 전에 설계

네 상품의 가격과 판매량으로 다음을 구합니다.

1. 상품별 매출
2. 전체 매출
3. 매출이 400,000 이상인 상품 이름
4. 매출이 높은 순서의 상품 이름

진행 순서:

- Prediction: 결과 개수와 첫 상품 매출 예측
- Build: 계산식 제안
- Evidence: 출력과 검증식 실행
- Interpretation: 어떤 상품을 우선 관리할지 근거 한 문장

````python
final_products = np.array(["노트", "펜", "가방", "텀블러"])
final_prices = np.array([3000, 1500, 50000, 20000])
final_quantities = np.array([120, 200, 15, 30])

print("상품 수:", final_products.size)
print("shape 일치:",
      final_products.shape == final_prices.shape == final_quantities.shape)
````

````python
final_revenue = final_prices * final_quantities
final_total = final_revenue.sum()
high_mask = final_revenue >= 400000
revenue_order = np.argsort(final_revenue)[::-1]

print("상품별 매출:", final_revenue)
print("전체 매출:", final_total)
print("40만 이상 상품:", final_products[high_mask])
print("매출 내림차순 상품:", final_products[revenue_order])
````

````python
assert final_revenue.tolist() == [360000, 300000, 750000, 600000]
assert final_total == 2010000
assert final_products[high_mask].tolist() == ["가방", "텀블러"]
assert final_products[revenue_order].tolist() == ["가방", "텀블러", "노트", "펜"]

print("모든 최종 검증을 통과했습니다.")
````
