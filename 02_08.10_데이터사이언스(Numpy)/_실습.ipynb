```python
#2차원 데이터 슬라이싱
regions = np.array(["서울", "부산", "대전"])
quarters = np.array(["Q1", "Q2", "Q3", "Q4"])
sales = np.array([
    [10, 12, 14, 16],
    [20, 18, 21, 25],
    [11, 13, 15, 17],
])

print("지역:", regions)
print("분기:", quarters)
print("매출표:\n", sales)
print("shape:", sales.shape)

print("부산·대전의 Q2~Q4:\n", sales[1:3, 1:4])
print("모든 지역의 Q2~Q3:\n", sales[:, 1:3])
print("서울 지역의 Q2~Q3:\n", sales[0:1, 1:3] )
print("서울 지역의 Q2~Q4:\n", sales[0:1, 1:4] )
print("서울 지역의 Q1~Q4:\n", sales[0:1, :4] )

sales[0:1]
```

```python
#1. 마스킹작업
years = np.arange(2010, 2020)
gdp = np.array([6610, 7637, 8127, 8885, 10385,
                12565, 14403, 12398, 8282, 10672])

mask = gdp > 10000
over_10000 = gdp > 10000

print("GDP:", gdp)
print("조건:", over_10000)
print("~조건(not):", ~over_10000)
print("조건 dtype:", over_10000.dtype)
print("같은 shape:", gdp.shape == over_10000.shape) ##항상 점검


mask_result = gdp[over_10000]
print(f'mask_result: {mask_result}')
print(f'mask_result revers: {~mask_result}')

print(\n)
##################################### 중요 : 컬럼 위치 전제 
mask = gdp > 10000
selected_years = years[mask] #--> selected_gdp 와 컬럼 위치가 같다는 전제 (True, False, ...)
selected_gdp = gdp[mask]
print("선택된 연도:", selected_years)
print("선택된 GDP :", selected_gdp)
print("모두 조건 만족?", np.all(selected_gdp > 10000))


```
