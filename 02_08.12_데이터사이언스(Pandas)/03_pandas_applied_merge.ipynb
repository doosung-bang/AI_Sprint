```python
import pandas as pd

orders = pd.read_csv("data/orders.csv")
customers = pd.read_csv("data/customers.csv")

orders["sales"] = orders["price"] * orders["quantity"]
orders["discount"] = orders["discount"].fillna(0) #    #### na 값 0 으로 채움
orders["final_sales"] = orders["sales"] * (1 - orders["discount"])
orders["order_date"] = pd.to_datetime(orders["order_date"])  #### 데이트 형식으로 변환(날짜 계산을 위해)
orders["region"] = orders["region"].str.strip()     #### 집계가 필요한 str 컬럼에대한 스페이스/탭/엔터 제거

logical_orders = (  #### 샘플링
    orders.loc[orders["record_status"].ne("ingestion_error")]  ##상태 값컬럼 에러가 아닌
    .drop_duplicates(subset="order_id", keep="first")      ###order_id에 대한 중복 제거 (first로 부터)
    .copy()                 ## 필터링한 데이터 복사
)

pd.Series({
    "logical_rows": len(logical_orders),  ###생성된 logical_orders 행수 확인
    "unique_order_id": logical_orders["order_id"].nunique(), ## order_id 기준 unique행 갯수
                                           ### nunique메서드는 선택된 축에 대해서 고유한 요소의 수를 구하는 메서드
})
```

두 표를 `customer_id`로 연결하면 아래 표가 어떻게 채워질지 먼저 예상하세요.

| customer_id | sales | membership |

```python
import pandas as pd

orders_small = pd.DataFrame({
    "customer_id": ["C01", "C02", "C03"],
    "sales": [50000, 30000, 70000],
})

customers_small = pd.DataFrame({
    "customer_id": ["C01", "C02", "C03"],
    "membership": ["VIP", "BASIC", "PLUS"],
})
```

```python
orders_small
```

```python
customers_small
```

```python
orders_small.merge(
    customers_small,
    on="customer_id",
    how="left"
)
orders_small_merge = orders_small.merge(
    customers_small,
    on="customer_id",
    how="left"
)

orders_small_merge
```

## Merge Mental Model

`주문 데이터 + 고객 데이터 + 공통 Key → 하나의 분석 가능한 데이터`

`how="left"`는 이번 분석의 기준인 주문을 유지하면서 찾을 수 있는 고객 정보를 옆에 붙입니다.
