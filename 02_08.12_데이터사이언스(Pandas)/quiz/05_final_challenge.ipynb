## 분석용 표 준비

아래 셀은 오늘 배운 내용으로 원본 판매 자료를 분석 가능한 상태로 준비합니다.

- 비어 있는 할인율은 0으로 채웁니다.
- 상품군 표기를 정리하고 날짜를 날짜 자료형으로 바꿉니다.
- `record_status`가 `verified`인 거래만 사용합니다.
- 판매 ID 한 개당 한 행을 남깁니다.
- 할인 전·후 판매금액을 계산합니다.

새로운 함수나 숨겨진 도우미를 쓰지 않습니다. 코드를 위에서 아래로 읽고 실행한 뒤 문제를 시작하세요.


```python
import pandas as pd

# 원본 파일은 그대로 두고, 아래에서 분석에 사용할 표를 따로 만듭니다.
sales_data = pd.read_csv("data/unseen.csv")
buyers = pd.read_csv("data/unseen_customers.csv")

clean_sales = sales_data.copy()
clean_sales["coupon_rate"] = clean_sales["coupon_rate"].fillna(0)
clean_sales["item_group"] = clean_sales["item_group"].str.strip().str.upper()
clean_sales["sold_at"] = pd.to_datetime(clean_sales["sold_at"])

# record_status가 verified인 거래만 남기고, sale_id 한 개당 한 행을 사용합니다.
clean_sales = clean_sales.loc[clean_sales["record_status"] == "verified"]
clean_sales = clean_sales.drop_duplicates("sale_id", keep="first").copy()

# 오늘 배운 열 계산으로 할인 전·후 판매금액을 준비합니다.
clean_sales["gross_sales"] = clean_sales["unit_price"] * clean_sales["units"]
clean_sales["final_sales"] = clean_sales["gross_sales"] * (1 - clean_sales["coupon_rate"])

clean_sales.head()
```
