## 4. 행과 열을 함께 선택하기

`loc`은 **라벨**, `iloc`은 **위치**를 기준으로 찾습니다.

```python
students.loc[0]
```

```python
students.iloc[0]
```

```python
students.loc[1]
```

```python
students.iloc[1]
```

`loc[0:2]`는 끝 라벨(컬럼) 2를 포함합니다. 몇 행과 몇 열이 나올지 예상해보세요.

```python
students.loc[0:3, ["name", "score"]] #위치기반 - 슬라이싱 숫자에 해당하는 위치를 가져옴
```

`iloc`의 위치 범위에서는 끝 위치가 포함되지 않습니다.

```python
students.iloc[0:3, 0:2] # 인덱스 기반 --> 인덱스 숫자 앞번호 까지만 데이터 가져오는 문법(python)
```

```python
# 첫 번째 학생의 정보를 loc으로 확인하세요.
# 처음 3명의 name과 score를 loc으로 선택하세요.
# iloc으로 첫 2행, 첫 2열을 선택하세요.
pass
print(students.head(1))
x = students.loc[0:2, ['name', 'score']]
print(x), type(x)
y = students.iloc[0:1, 0:2]
print(y)
```

# 5. 조건에 맞는 데이터 찾기

80점 이상인 학생을 바로 고르기 전에, 각 학생이 조건에 맞는지 먼저 확인합니다.

실행하면 숫자, 이름, True/False 중 무엇이 나올까요?

```python
students["score"] >= 80
```

### Boolean Mask

조건을 적용하면 각 행에 대해 `True / False`가 만들어집니다.

`조건 → 각 행 비교 → True / False`

```python
mask = students["score"] >= 80
mask
```

이제 `True`인 행만 골라낼 수 있을까요?

```python
students[mask]
```

```python
students[students["score"] >= 80]
```

숫자 비교와 문자열 비교도 `조건 → True/False → 선택`이라는 흐름은 같습니다.

```python
students["city"] == "서울"
```

```python
students[students["city"] == "서울"]
```

서울 학생이면서 점수가 80점 이상인 학생을 찾으려면 어떤 두 조건이 필요할까요?

```python
students[
    (students["city"] == "서울")
    & (students["score"] >= 80)
]
```

`&`는 그리고, `|`는 또는입니다. 여러 조건은 각각 괄호로 묶습니다.

```python
students[
    (students["city"] == "서울")
    | (students["city"] == "부산")
]
```

### Filtering 연습

한 문제씩 조건을 만들고, 그 조건이 `True`인 행을 선택하세요.

```python
# 점수가 85점 이상인 학생을 찾으세요.
pass
print(students)
mask = students['score'] >= 85
students[mask]
```

```python
# 부산 학생만 찾으세요.
pass
print(students[students['city']=='부산'])
students[students['city']=='부산']
```

```python
# 부산 학생이면서 점수가 85점 이상인 학생을 찾으세요.
pass

students[(students['city']=='부산') & (students['score']>=85)]
```

```python
# 서울 또는 부산에 사는 학생을 찾으세요.
pass

print(students[(students['city']=='서울')])
print(students[(students['city']=='부산')])
students[ (students['city']=='서울') | (students['city']=='부산') ]
```

# 7. 작은 판매 데이터

가격과 수량은 있지만 주문금액은 없습니다. 기존 열로 새 값을 만들어봅니다.

```python
sales = pd.DataFrame({
    "product": ["키보드", "마우스", "노트"],
    "price": [50000, 25000, 3000],
    "quantity": [1, 2, 3],
})
sales
```

각 상품의 `가격 × 수량`을 손으로 먼저 계산한 뒤 아래 셀의 결과를 예상하세요.

```python
sales["price"] * sales["quantity"]
```

```python
sales["sales"] = sales["price"] * sales["quantity"]
sales
```

```python
sales["discount"] = [0.1, 0.0, 0.2]
sales
```

```python
# 할인 전 sales와 discount를 이용해 discount_amount 열을 만들어보세요.
pass

discount_amount = sales['price'] * sales['quantity'] * sales['discount']
sales['discount_amount'] =discount_amount
sales
```
