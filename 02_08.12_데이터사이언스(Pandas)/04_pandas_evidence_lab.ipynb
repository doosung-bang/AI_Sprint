# Pandas를 이렇게 기억합니다

| 질문 | 떠올릴 수 있는 방법 |
|---|---|
| 무엇이 들어 있지? | `head`, `shape`, `info` |
| 필요한 부분은? | 열과 행 선택 |
| 어떤 조건의 데이터가 필요하지? | Boolean Filtering |
| 원하는 값이 없으면? | 열 계산과 새 열 |
| 이 데이터를 믿어도 되나? | `isna`, `duplicated`, 문자열 확인 |
| 그룹별 차이는? | `groupby`, `agg` |
| 다른 파일의 정보가 필요하면? | `merge` |
| 그래서 무엇을 말할 수 있지? | Evidence |
