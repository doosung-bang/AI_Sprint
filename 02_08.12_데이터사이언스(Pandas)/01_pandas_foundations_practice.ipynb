# Mission 4 — 조건에 맞는 상품 찾기

완성된 코드를 떠올리기 전에, 어떤 열에 어떤 조건을 만들어야 하는지 먼저 생각하세요.

```python
# 가격이 30,000원 이상인 상품만 선택하세요.
pass
products['price']>=30000
print( products[products['price']>=30000] )
print( "총 행수", (products['price']>=30000).sum() )
```
