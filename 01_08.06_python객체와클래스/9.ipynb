## 1. 하나의 리스트를 두 변수가 참조하면?


```python
subjects1 = ['Python']
subjects2 = subjects1

subjects2.append('SQL')

print('subjects1:', subjects1)
print('subjects2:', subjects2)
print('같은 객체인가?', subjects1 is subjects2)
```

### 실행 결과 해석

subjects2에만 SQL을 추가한 것처럼 보이지만
두 변수는 같은 리스트 객체를 참조한다.
따라서 subjects1에서도 SQL을 확인할 수 있다.


## 2. 클래스에서도 같은 문제가 생길까?


```python
class SharedStudent:
    def __init__(self, name, subjects=[]):
        self.name = name
        self.subjects = subjects

    def add_subject(self, subject):
        self.subjects.append(subject)

    def show_info(self):
        return f'{self.name}: {self.subjects}'


student1 = SharedStudent('민수')
student2 = SharedStudent('지수')

print()
print('과목 추가 전')
print(student1.show_info())
print(student2.show_info())
```

### 예측과 판단

실행하기 전에 결과를 예측하세요.


```python
student1.add_subject('Python')

print()
print('민수에게 Python 추가 후')
print(student1.show_info())
print(student2.show_info())
```

student1에만 과목을 추가했지만
student2의 과목 목록에도 Python이 나타난다.


## 3. 정말 같은 리스트인지 확인하기


```python
print()
print('student1 목록 ID:', id(student1.subjects))
print('student2 목록 ID:', id(student2.subjects))
print(
    '같은 목록인가?',
    student1.subjects is student2.subjects
)
```

subjects=[]는 객체를 생성할 때마다
새로운 리스트를 만드는 코드가 아니다.

함수가 정의될 때 만들어진 하나의 리스트가
기본값으로 계속 사용된다.


## 4. None을 사용해 문제 해결하기


```python
class Student:
    def __init__(self, name, subjects=None):
        self.name = name

        if subjects is None:
            self.subjects = []
        else:
            self.subjects = list(subjects)

    def add_subject(self, subject):
        self.subjects.append(subject)

    def show_info(self):
        return f'{self.name}: {self.subjects}'


student3 = Student('민수')
student4 = Student('지수')

student3.add_subject('Python')

print()
print('수정한 클래스의 결과')
print(student3.show_info())
print(student4.show_info())

print()
print(
    '같은 목록인가?',
    student3.subjects is student4.subjects
)
```

subjects가 전달되지 않으면
객체를 만들 때마다 새로운 리스트를 저장한다.


## 5. 전달받은 리스트도 복사하기


```python
original_subjects = ['Python']
student5 = Student('현우', original_subjects)

original_subjects.append('SQL')

print()
print('원본 목록:', original_subjects)
print('학생의 목록:', student5.subjects)

print(
    '같은 목록인가?',
    original_subjects is student5.subjects
)
```

### 실행 결과 해석

self.subjects = list(subjects)는
전달받은 리스트의 내용을 복사해 새로운 리스트를 만든다.
따라서 외부 리스트를 변경해도 학생의 목록은 바뀌지 않는다.


## 6. 인스턴스별 상태 확인하기


```python
students = [
    Student('민수', ['Python']),
    Student('지수', ['SQL']),
    Student('현우')
]

print()
print('학생별 수강 과목')

for student in students:
    print(student.show_info())
```

각 Student 객체는 독립적인 subjects 리스트를 가진다.


## 핵심 정리


피해야 하는 코드

def __init__(self, subjects=[]):
    self.subjects = subjects

권장하는 코드

def __init__(self, subjects=None):
    if subjects is None:
        self.subjects = []
    else:
        self.subjects = list(subjects)

list, dict, set과 같은 가변 객체를
함수나 메서드의 기본값으로 직접 사용하지 않는다.
