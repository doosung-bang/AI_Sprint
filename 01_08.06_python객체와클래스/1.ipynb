## 6. 클래스로 자동차 설계하기


```python
class Car:
    """자동차 한 대의 데이터와 기능을 표현하는 클래스."""

    def __init__(self, company, color, horsepower, price):
        # __init__은 객체가 만들어질 때 자동으로 실행된다.
        # self는 현재 생성되는 객체 자신을 의미한다.
        self.company = company
        self.color = color
        self.horsepower = horsepower
        self.price = price

    def __str__(self):
        # print(객체)를 실행할 때 사용하는 문자열
        # 일반적
        return (
            f'{self.company} / {self.color} / '
            f'{self.horsepower}마력 / 가격 {self.price:,}원'
        )

    def __repr__(self):
        # 리스트 등에서 객체를 확인할 때 사용하는 문자열
        # 개발자 연구자 볼려고 쓰는 친구
        return (
            f"Car(company='{self.company}', "
            f"color='{self.color}', "
            f"horsepower={self.horsepower}, "
            f"price={self.price})"
        )

    def apply_discount(self, discount_rate):
        """할인율을 전달받아 가격을 변경한다."""

        if not 0 <= discount_rate <= 1:
            raise ValueError('할인율은 0 이상 1 이하여야 합니다.')

        self.price = round(
            self.price * (1 - discount_rate)
        )

    def is_high_performance(self):
        """300마력 이상인지 확인한다."""

        return self.horsepower >= 200
```

## 8. 객체를 리스트로 관리하기


```python
car_list = [car1, car2, car3]

print()
print('자동차 목록')
print(car_list)

print()
print('자동차별 정보')

for car in car_list:
    print(car)
```

## 9. 객체의 기능 사용하기


```python
print()
print('고성능 자동차 확인')

for car in car_list:
    if car.is_high_performance():
        print(f'{car.company}: 고성능 자동차')
    else:
        print(f'{car.company}: 일반 자동차')
```
