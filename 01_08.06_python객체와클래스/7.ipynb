# Chapter 07 - @property
파이썬 클래스

학습 목표
1. @property의 역할을 설명할 수 있다.
2. 속성처럼 값을 읽고 변경할 수 있다.
3. Setter에서 잘못된 값을 검사할 수 있다.


```python
class Student:
    def __init__(self, name, score):
        self.name = name

        # Setter를 통해 최초 점수를 저장한다.
        self.score = score

    @property
    def score(self):
        """점수를 읽을 때 실행된다."""

        return self.__score

    @score.setter
    def score(self, value):
        """점수를 저장하거나 변경할 때 실행된다."""

        if not isinstance(value, int):
            raise TypeError('점수는 정수여야 합니다.')

        if not 0 <= value <= 100:
            raise ValueError('점수는 0부터 100 사이여야 합니다.')

        self.__score = value
```

## 1. 객체 생성하기


```python
student = Student('민수', 80)

print('학생:', student.name)
print('현재 점수:', student.score)
```

student.score를 읽으면
@property가 적용된 score()가 실행된다.


## 2. 속성처럼 값 변경하기


```python
student.score = 95

print()
print('변경된 점수:', student.score)
```

student.score = 95를 실행하면
@score.setter가 적용된 score()가 실행된다.


## 3. 잘못된 범위의 값 전달하기


```python
print()

try:
    student.score = 150
except ValueError as error:
    print('점수 변경 실패:', error)

print('현재 점수:', student.score)
```

잘못된 값은 저장되지 않고 95점이 유지된다.


## 4. 잘못된 자료형 전달하기


```python
print()

try:
    student.score = '백 점'
except TypeError as error:
    print('점수 변경 실패:', error)

print('현재 점수:', student.score)
```

## 5. 객체를 만들 때도 검증될까?


```python
print()

try:
    wrong_student = Student('지수', -10)
except ValueError as error:
    print('학생 생성 실패:', error)
```

### 실행 결과 해석

__init__()에서 self.score = score를 사용했기 때문에
객체를 만들 때 전달한 점수도 Setter의 검사를 받는다.


## 6. Getter, Setter 방식과 비교하기


일반 Getter, Setter 방식

student.get_score()
student.set_score(90)

@property 방식

student.score
student.score = 90

외부에서는 일반 속성처럼 사용하지만
내부에서는 값을 읽고 변경하는 메서드가 실행된다.


## 핵심 정리


값을 읽을 때
@property

값을 변경할 때
@속성이름.setter

@property를 사용하면 속성처럼 간단하게 접근하면서
내부에서는 자료형과 값의 범위를 검사할 수 있다.
