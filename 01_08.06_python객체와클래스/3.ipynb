# Chapter 03 - 인스턴스·클래스·정적 메서드
파이썬 클래스

학습 목표
1. self와 cls의 차이를 설명할 수 있다.
2. 필요한 정보에 따라 적절한 메서드를 선택할 수 있다.
3. 클래스 메서드로 클래스 변수를 안전하게 변경할 수 있다.


```python
class Car:
    """자동차의 회사와 가격을 표현하는 클래스."""

    # 모든 Car 객체가 공유하는 클래스 변수
    price_raise_rate = 1.0

    def __init__(self, company, price):
        self.company = company
        self.price = price

    # 인스턴스 메서드
    # 특정 자동차 객체의 정보가 필요하다.
    def show_info(self):
        return f'{self.company}: {self.price:,}원'

    # 인스턴스 메서드
    def get_raised_price(self):
        return self.price * self.price_raise_rate

    # 클래스 메서드
    # 특정 자동차가 아니라 Car 클래스의 정보가 필요하다.
    # @classmethod
    # def set_price_raise_rate(cls, rate):
    #     if not isinstance(rate, (int, float)):
    #         raise TypeError('가격 인상 배율은 숫자여야 합니다.')
    #
    #     if rate < 1:
    #         raise ValueError('가격 인상 배율은 1 이상이어야 합니다.')
    #
    #     cls.price_raise_rate = rate

    # 정적 메서드
    # 특정 객체와 클래스 변수 모두 필요하지 않다.
    @staticmethod
    def is_valid_company_name(company):
        return (
            isinstance(company, str)
            and len(company.strip()) >= 2
        )
```

## 1. 자동차 객체 생성하기


```python
car1 = Car('BMW', 5000)
car2 = Car('Audi', 6000)

print(car1.show_info())
print(car2.show_info())
```

## 2. 인스턴스 메서드


```python
print()
print('가격 인상 전')

print(f'{car1.company}: {car1.get_raised_price():,.0f}원')
print(f'{car2.company}: {car2.get_raised_price():,.0f}원')
```

### 실행 결과 해석

get_raised_price()는 현재 자동차의 price가 필요하다.
따라서 인스턴스 메서드로 작성한다.

인스턴스 메서드의 첫 번째 매개변수는 self이다.
self는 메서드를 호출한 현재 객체를 의미한다.


## 3. 클래스 변수를 직접 변경하면 생기는 문제


다음처럼 클래스 변수를 직접 변경할 수 있다.

Car.price_raise_rate = -10

하지만 잘못된 값도 그대로 저장될 수 있다.
변경 규칙을 클래스 메서드에 넣으면
값을 저장하기 전에 검사할 수 있다.


## 4. 클래스 메서드


```python
# Car.set_price_raise_rate(1.2)

car1.price_raise_rate = 2

print()
print('가격 인상 후')
print('공통 인상 배율:', Car.price_raise_rate)

print(f'{car1.company}: {car1.get_raised_price():,.0f}원')
print(f'{car2.company}: {car2.get_raised_price():,.0f}원')
```

### 실행 결과 해석

set_price_raise_rate()는 특정 자동차의 정보가 필요하지 않다.
모든 자동차가 공유하는 클래스 변수를 변경한다.

클래스 메서드의 첫 번째 매개변수는 cls이다.
여기서 cls는 Car 클래스를 의미한다.


## 5. 잘못된 가격 인상 배율 처리하기


```python
print()

try:
    Car.set_price_raise_rate(0.5)
except ValueError as error:
    print('가격 인상 실패:', error)

print('현재 인상 배율:', Car.price_raise_rate)
```

잘못된 값은 저장되지 않고 기존 값이 유지된다.


## 6. 정적 메서드


```python
print()
print('회사 이름 검사')

print('BMW:', Car.is_valid_company_name('BMW'))
print('A:', Car.is_valid_company_name('A'))
print('빈 문자열:', Car.is_valid_company_name(''))
print('숫자:', Car.is_valid_company_name(100))
```

is_valid_company_name()은 전달받은 문자열만 검사한다.
특정 자동차 객체나 클래스 변수를 사용하지 않는다.

다만 자동차 회사 이름을 검사하는 기능이므로
Car 클래스 안에 정적 메서드로 배치했다.


## 7. 세 가지 메서드 선택 기준


특정 객체의 정보가 필요하다.
-> 인스턴스 메서드
-> 첫 번째 매개변수: self

클래스가 공유하는 정보가 필요하다.
-> 클래스 메서드
-> 첫 번째 매개변수: cls
-> @classmethod 사용

객체와 클래스의 정보가 모두 필요하지 않다.
-> 정적 메서드
-> self와 cls를 받지 않음
-> @staticmethod 사용
