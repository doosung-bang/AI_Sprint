## 1. 파이썬의 기본 자료형도 클래스다


```python
print(int)
print(float)
print(str)
print(list)

print()
print(type(10))
print(type(3.14))
print(type('Python'))
print(type([1, 2, 3]))
```

int, float, str, list는 클래스이다.
10, 3.14, 'Python', [1, 2, 3]은
각각의 클래스로 만든 객체이다.


## 2. 객체와 클래스의 관계 확인하기


```python
print()
print(isinstance(10, int))
print(isinstance(3.14, float))
print(isinstance('Python', str))
print(isinstance([1, 2, 3], list))
print(isinstance([1, 2, 3], dict))

```

isinstance(값, 클래스)는
값이 해당 클래스로 만든 객체인지 확인한다.


## 3. 연산자와 특수 메서드


```python
number = 10
print(type(number))

print()
print('덧셈')
print(number + 5)
print(number.__add__(5))

print()
print('곱셈')
print(number * 3)
print(number.__mul__(3))

print()
print('논릿값 변환')
print(bool(number))
print(number.__bool__())
```

다음 표현은 서로 관련이 있다.

number + 5
-> number.__add__(5)

number * 3
-> number.__mul__(3)

bool(number)
-> number.__bool__()

특수 메서드의 이름은 앞뒤에 밑줄 두 개가 붙는다.
일반 코드에서는 특수 메서드를 직접 호출하기보다
+, *, bool()과 같은 파이썬 문법을 사용한다.


## 4. 출력 방법을 정의하지 않은 객체


```python
class BasicProduct:
    def __init__(self, name, price):
        self.name = name
        self.price = price


basic_product = BasicProduct('Keyboard', 35000)

print()
print(basic_product)
```

__str__이나 __repr__을 정의하지 않으면
객체의 종류와 위치를 나타내는 정보가 출력된다.
사람에게 필요한 상품 정보는 확인하기 어렵다.


## 5. __str__과 __repr__ 정의하기


```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        """사용자가 읽기 편한 문자열을 반환한다."""

        return f'{self.name}: {self.price:,}원'

    def __repr__(self):
        """개발자가 객체를 확인하기 좋은 문자열을 반환한다."""

        return (
            f"Product(name='{self.name}', "
            f'price={self.price})'
        )


product1 = Product('Keyboard', 35000)
product2 = Product('Mouse', 18000)
```

## 6. __str__ 확인하기


```python
print()
print(product1)
print(str(product1))
```

print(product1)과 str(product1)은
Product의 __str__()을 사용한다.


## 7. __repr__ 확인하기


```python
print()
print(repr(product1))
print([product1, product2])
```

repr(product1)은 __repr__()을 사용한다.
리스트 안에 객체를 넣어 출력할 때도
일반적으로 각 객체의 __repr__()이 사용된다.


## 8. 두 표현의 목적 비교하기


```python
print()
print('사용자용:', str(product1))
print('개발자용:', repr(product1))
```

__str__
사용자에게 보여 줄 읽기 편한 표현

__repr__
개발자가 객체의 상태를 확인하기 좋은 표현
