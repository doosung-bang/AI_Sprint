## 1. 부모 클래스 만들기


```python
class Animal:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f'제 이름은 {self.name}입니다.'

    def sound(self):
        return '동물이 소리를 냅니다.'


animal = Animal('동물')

print(animal.introduce())
print(animal.sound())
```

## 2. 부모 메서드 상속받기


```python
class BasicDog(Animal):
    pass


basic_dog = BasicDog('초코')

print()
print(basic_dog.introduce())
print(basic_dog.sound())
```

BasicDog에는 introduce()와 sound()가 없다.
파이썬은 BasicDog에서 메서드를 찾지 못하면
부모 클래스인 Animal에서 메서드를 찾는다.


## 3. 부모 메서드 재정의하기


```python
class Dog(Animal):
    def sound(self):
        return '멍멍'


class Cat(Animal):
    def sound(self):
        return '야옹'


dog = Dog('콩이')
cat = Cat('나비')

print()
print(dog.introduce())
print(dog.sound())
print(cat.introduce())
print(cat.sound())
```

```python
class Hoho(object):
    pass
```

Dog와 Cat은 Animal의 sound()를
같은 이름으로 다시 정의했다.
이를 메서드 오버라이딩이라고 한다.


## 4. 다형성 확인하기


```python
animals = [
    Animal('동물'),
    Dog('초코'),
    Cat('나비')
]

print()
print('동물 소리 확인')

for current_animal in animals:
    print(
        current_animal.name,
        '→',
        current_animal.sound()
    )
```

모든 객체에 sound()를 똑같이 호출했다.
하지만 객체의 종류에 따라 결과가 다르다.

같은 메서드 호출이 객체에 따라 다르게 동작하는 것을
다형성이라고 한다.


## 5. super()로 부모 메서드 확장하기


```python
class LoudDog(Dog):
    def sound(self):
        parent_sound = super().sound()

        return parent_sound + '!!!'


loud_dog = LoudDog('두부')

print()
print(loud_dog.sound())
```

super().sound()
-> 부모 클래스인 Dog의 sound()를 호출한다.

부모 기능을 완전히 바꾸지 않고
추가 기능을 연결할 때 super()를 사용할 수 있다.


## 6. 생성자에서 super() 사용하기


```python
class Bird(Animal):
    def __init__(self, name, can_fly):
        # name 저장은 부모 클래스에 맡긴다.
        super().__init__(name)

        # Bird에만 필요한 속성을 추가한다.
        self.can_fly = can_fly

    def sound(self):
        return '짹짹'

    def show_flight(self):
        if self.can_fly:
            return f'{self.name}: 날 수 있습니다.'

        return f'{self.name}: 날 수 없습니다.'


bird = Bird('참새', True)

print()
print(bird.introduce())
print(bird.sound())
print(bird.show_flight())
```

super().__init__(name)
-> Animal의 __init__()을 실행한다.

부모가 담당하는 name은 부모에게 맡기고,
Bird는 자신에게 필요한 can_fly를 추가한다.


## 7. 객체와 클래스의 관계 확인하기


```python
print()
print(isinstance(dog, Dog))
print(isinstance(dog, Animal))
print(isinstance(cat, Dog))
```

Dog는 Dog 클래스의 객체이면서
Animal을 상속받은 자식 클래스의 객체이기도 하다.
