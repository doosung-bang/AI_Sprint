# Chapter 02 - self, 인스턴스 변수, 클래스 변수
파이썬 클래스

학습 목표
1. self가 현재 객체를 의미한다는 것을 설명할 수 있다.
2. 인스턴스 변수와 클래스 변수의 차이를 설명할 수 있다.
3. 여러 객체가 독립적인 상태와 공유 상태를 가지는 모습을 확인할 수 있다.


```python
class Car:
    """자동차 한 대의 정보를 표현하는 클래스."""

    # 클래스 변수
    # 모든 Car 객체가 함께 사용하는 값이다.
    total_created = 0

    def __init__(self, company, color, price):
        # 인스턴스 변수
        # 자동차 객체마다 서로 다른 값을 가진다.
        self.company = company
        self.color = color
        self.price = price

        Car.total_created += 1

    def show_info(self):
        return (
            f'{self.company} / {self.color} / '
            f'{self.price:,}원'
        )
```

## 1. 객체 생성하기


```python
car1 = Car('Ferrari', 'White', 8000)
car2 = Car('BMW', 'Black', 5000)
car3 = Car('Audi', 'Silver', 6000)
car4 = Car('Audi', 'Silver', 18)

print(car1.color)
print(car2.color)
print(car1.show_info())
print(car2.show_info())
print(car3.show_info())

print()
print()
print(car1)

```

## 2. self의 의미 확인하기


```python
print()
print('두 가지 메서드 호출 방식')

print(car1.show_info())
print(Car.show_info(car1))
```

위 두 코드는 같은 결과를 반환한다.

car1.show_info()
-> 파이썬이 car1을 self에 자동으로 전달한다.

Car.show_info(car1)
-> 호출하는 사람이 car1을 직접 전달한다.


## 3. 객체가 서로 다른지 확인하기


```python
print()
print('객체 ID 확인')

print('car1:', id(car1))
print('car2:', id(car2))
print('car3:', id(car3))

print()
print('car1 is car2:', car1 is car2)
```

car1, car2, car3는 모두 Car로 만들었지만
서로 다른 객체이다.


## 4. 인스턴스 변수의 독립성 확인하기


```python
print()
print('가격 변경 전')
print('car1:', car1.price)
print('car2:', car2.price)

car1.price = 7500

print()
print('car1 가격 변경 후')
print('car1:', car1.price)
print('car2:', car2.price)
```

car1의 price만 변경된다.
car2의 price에는 영향을 주지 않는다.


## 5. 인스턴스 변수 확인하기


```python
print()
print('인스턴스에 저장된 값')

print('car1:', car1.__dict__)
print('car2:', car2.__dict__)
```

### 실행 결과 해석

__dict__를 보면 각 객체에 직접 저장된
인스턴스 변수를 확인할 수 있다.


## 6. 클래스 변수 확인하기


```python
print()
print('생성한 자동차 수:', Car.total_created)

print('Car에서 접근:', Car.total_created)
print('car1에서 접근:', car1.total_created)
print('car2에서 접근:', car2.total_created)
car1.total_created = 0
print('car1에서 접근:', car1.total_created)
```

total_created는 Car 클래스에 저장된 클래스 변수이다.
인스턴스를 통해 읽을 수도 있지만,
클래스 변수라는 사실을 명확히 보여주기 위해
Car.total_created로 접근하는 것이 좋다.


## 7. 객체를 추가로 생성하기


```python
car4 = Car('Volvo', 'Blue', 4500)

print()
print('car4 생성 후:', Car.total_created)
```

새로운 Car 객체를 만들면
모든 객체가 공유하는 total_created가 증가한다.


## 8. 클래스 변수와 같은 이름을 인스턴스에 저장하면?


```python
print()
print('같은 이름을 저장하기 전')
print('car1.total_created:', car1.total_created)
print('Car.total_created:', Car.total_created)

car1.total_created = 100

print()
print('car1.total_created = 100 실행 후')
print('car1.total_created:', car1.total_created)
print('car2.total_created:', car2.total_created)
print('Car.total_created:', Car.total_created)
```

car1.total_created = 100은 클래스 변수를 변경한 것이 아니다.
car1 객체에 같은 이름의 인스턴스 변수를 새로 만든 것이다.

그래서 __init__()에서는 다음처럼 작성했다.

Car.total_created += 1

self.total_created += 1로 작성하면 객체에 같은 이름의
인스턴스 변수가 생겨 클래스의 공유 값을 바꾸지 못할 수 있다.


```python
print()
print('car1에 직접 저장된 값:', car1.__dict__)
```

## 9. 같은 이름의 인스턴스 변수 제거하기


```python
del car1.total_created

print()
print('인스턴스 변수 제거 후')
print('car1.total_created:', car1.total_created)
print('Car.total_created:', Car.total_created)
```

car1에서 total_created를 찾지 못하면
파이썬은 Car 클래스에서 같은 이름을 찾는다.


## 10. 객체와 클래스의 관계 확인하기


```python
print()
print(isinstance(car1, Car))
print(isinstance(car2, Car))
```

isinstance(객체, 클래스)는
해당 객체가 그 클래스로 만든 객체인지 확인한다.


## 핵심 정리


self

인스턴스 변수
