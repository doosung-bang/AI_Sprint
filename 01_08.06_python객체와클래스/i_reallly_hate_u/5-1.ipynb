활동별 테스트
아래 코드는 수정하지 않습니다.


```python
def test_activity_1():
    box1 = AppleBox(10)
    box2 = AppleBox(4)

    result = box1 - box2

    print('box1:', box1)
    print('box2:', box2)
    print('결과:', result)

    assert isinstance(result, AppleBox), (
        '결과는 AppleBox 객체여야 합니다.'
    )
    assert result.apple_count == 6, (
        '10개에서 4개를 빼면 6개가 되어야 합니다.'
    )
    assert box1.apple_count == 10, (
        '기존 box1을 변경하면 안 됩니다.'
    )
    assert box2.apple_count == 4, (
        '기존 box2를 변경하면 안 됩니다.'
    )

    empty_result = AppleBox(3) - AppleBox(10)

    assert empty_result.apple_count == 0, (
        '계산 결과가 음수이면 0개가 되어야 합니다.'
    )


def test_activity_2():
    box1 = AppleBox(5)
    box2 = AppleBox(5)
    box3 = AppleBox(2)

    print('box1 == box2:', box1 == box2)
    print('box1 == box3:', box1 == box3)

    assert box1 == box2, (
        '사과 개수가 같으면 True여야 합니다.'
    )
    assert box1 != box3, (
        '사과 개수가 다르면 같지 않아야 합니다.'
    )


def test_activity_3():
    box1 = AppleBox(10)
    box2 = AppleBox(4)
    box3 = AppleBox(10)

    print('box1 > box2:', box1 > box2)
    print('box2 > box1:', box2 > box1)

    assert box1 > box2, (
        '10개는 4개보다 많아야 합니다.'
    )
    assert not box2 > box1, (
        '4개는 10개보다 많지 않습니다.'
    )
    assert not box1 > box3, (
        '개수가 같으면 더 많다고 판단하면 안 됩니다.'
    )


def test_activity_4():
    box1 = AppleBox(7)
    empty_box = AppleBox()

    print('box1의 사과 개수:', len(box1))
    print('빈 상자의 사과 개수:', len(empty_box))

    assert len(box1) == 7
    assert len(empty_box) == 0
```
