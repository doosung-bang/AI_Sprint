```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def __str__(self):
        # TODO
        # 사용자가 읽기 편한 문자열을 반환하세요.
        #
        # 출력 형식:
        # 야한책 - 김 / 20,000원

        return f'{self.title} - {self.author} / {self.price:,}원'

    def __repr__(self):
        # TODO
        # 개발자가 객체의 상태를 확인하기 좋은
        # 문자열을 반환하세요.
        #
        # 출력 형식:
        # Book(title='야한책', author='김', price=20000)

        return (
            f'Book(title={self.title!r}, '
            f'author={self.author!r}, price={self.price})'
        )
```

테스트 코드
아래 코드는 수정하지 않습니다.


```python
try:
    book1 = Book('야한책', '김', 20000)
    book2 = Book('야한책2', '이', 25000)

    # --------------------------------------------------------
    # 1. __str__ 테스트
    # --------------------------------------------------------

    print()
    print('1. 사용자가 읽기 편한 표현')
    print(book1)

    assert str(book1) == (
        '파이썬 첫걸음 - 김코딩 / 20,000원'
    ), '__str__()의 반환 문자열을 확인하세요.'

    print('__str__ 테스트 통과')

    # --------------------------------------------------------
    # 2. __repr__ 테스트
    # --------------------------------------------------------

    print()
    print('2. 개발자가 확인하기 좋은 표현')
    print(repr(book1))

    assert repr(book1) == (
        "Book(title='파이썬 첫걸음', "
        "author='김코딩', price=20000)"
    ), '__repr__()의 반환 문자열을 확인하세요.'

    print('__repr__ 테스트 통과')

    # --------------------------------------------------------
    # 3. 리스트 출력 테스트
    # --------------------------------------------------------

    print()
    print('3. 책 목록 출력')

    books = [book1, book2]

    print(books)

    assert repr(books) == (
        "[Book(title='야한책', "
        "author='김', price=20000), "
        "Book(title='야한책2', "
        "author='이', price=25000)]"
    ), '리스트에서는 각 객체의 __repr__()이 사용됩니다.'

    print('리스트 출력 테스트 통과')

    # --------------------------------------------------------
    # 4. 원본 데이터 유지 확인
    # --------------------------------------------------------

    assert book1.title == '야한책'
    assert book1.author == '김'
    assert book1.price == 20000

    print()
    print('완료')

except (AssertionError, NotImplementedError, TypeError, ValueError) as error:
    print()
    print('학생 활동')
    print('확인할 내용:', error)
```
