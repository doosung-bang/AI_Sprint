파이썬 클래스
한 객체가 다른 객체를 부품처럼 가지는 관계


1. 먼저 두 역할을 나눠 봅니다.


```python
class PowerSource:
    """자동차를 움직이게 하는 동력원"""

    def __init__(self, source_type, power):
        self.source_type = source_type
        self.power = power
        self.is_running = False

    def start(self):
        self.is_running = True
        return f'{self.source_type} 동력원을 시작합니다.'

    def stop(self):
        self.is_running = False
        return f'{self.source_type} 동력원을 멈춥니다.'

    def __str__(self):
        return f'{self.source_type} / 출력 {self.power}'


class Car:
    """동력원 객체를 부품으로 가지는 자동차"""

    def __init__(self, company, model, power_source):
        self.company = company
        self.model = model
        self.power_source = power_source

    def start(self):
        return f'{self.model}: {self.power_source.start()}'

    def stop(self):
        return f'{self.model}: {self.power_source.stop()}'

    def __str__(self):
        return f'{self.company} {self.model} ({self.power_source})'
```

2. 자동차 안에 동력원 객체를 넣습니다.


```python
gasoline_source = PowerSource('가솔린', 200)
electric_source = PowerSource('전기', 300)

avante = Car('Hyundai', 'Avante', gasoline_source)
ioniq = Car('Hyundai', 'Ioniq', electric_source)

print(avante)
print(ioniq)
print(avante.start())
print(ioniq.start())
print()
```

3. 상속 관계와 포함 관계를 출력으로 구분합니다.


```python
print(isinstance(avante, Car))
print(isinstance(avante, PowerSource))
print(isinstance(avante.power_source, PowerSource))
```

Car is a PowerSource?  -> 거짓

4. Car를 수정하지 않고 부품만 바꿔 재사용합니다.
