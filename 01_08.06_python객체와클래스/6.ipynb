## 1. Student 클래스와 속성 이름 관례


```python
class Student:
    def __init__(self, name, school, score):
        self.name = name          # 공개 속성
        self._school = school     # 내부용이라는 관례
        self.__score = score      # 이름 맹글링 적용

    # Getter
    def get_score(self):
        return self.__score

    # Setter
    def set_score(self, score):
        if not isinstance(score, int):
            raise TypeError('점수는 정수여야 합니다.')

        if not 0 <= score <= 100:
            raise ValueError('점수는 0부터 100 사이여야 합니다.')

        self.__score = score


student = Student('민수', '파이썬고등학교', 80)
```

## 2. 공개 속성


```python
print('이름:', student.name)

student.name = '지수'

print('변경된 이름:', student.name)
```

## 3. 밑줄 하나를 사용한 속성


```python
print()
print('학교:', student._school)
```

밑줄 하나는 접근을 막는 문법이 아니다.
내부에서 사용할 목적이라는 개발자 간의 관례이다.


## 4. 밑줄 두 개를 사용한 속성


```python
print()

try:
    print(student.__score)
except AttributeError as error:
    print('직접 접근 실패:', error)

print()
print('객체에 저장된 값:', student.__dict__)

print(
    '변경된 내부 이름으로 접근:',
    student._Student__score
)
```

### 실행 결과 해석

__score는 다음과 비슷한 이름으로 변경된다.
_Student__score

이를 이름 맹글링(Name Mangling)이라고 한다.
따라서 완전히 접근할 수 없는 private 속성은 아니다.
위 출력은 이를 증명하기 위한 관찰이다.
실제 코드에서는 변경된 이름으로 직접 접근하지 않는다.


## 5. Getter로 값 읽기


```python
print()
print('현재 점수:', student.get_score())
```

## 6. Setter로 값 변경하기


```python
student.set_score(95)

print('변경된 점수:', student.get_score())
```

## 7. 잘못된 값 전달하기


```python
print()

try:
    student.set_score(150)
except ValueError as error:
    print('점수 변경 실패:', error)

print('현재 점수:', student.get_score())
```

## 8. Getter와 Setter의 불편함


일반 속성


```python
student.name = '현우'
print(student.name)
```

Getter와 Setter


```python
student.set_score(90)
print(student.get_score())
```
