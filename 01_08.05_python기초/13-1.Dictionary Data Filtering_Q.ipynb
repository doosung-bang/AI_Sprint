## 13. Dictionary Data Filtering💻
***
#### **`연습예제`** 💻

`아래와 같은` **`딕셔너리`** `구조에서` **`Value 값이 25 이상`** `필터링 후 출력하세요.` **`다양한 방법`** `으로 코딩하세요.`



```python
d = {'a': 8, 'b': 33, 'c': 15, 'd': 26, 'e': 12, 'f': 120}
출력 결과 : {'b': 33, 'd': 26, 'f': 120}
```

#### **`코딩영역`**

````python
# 코딩 영역 
#import this

### 딕션너리: key 와 value의 쌍으로 이루어진 리스트
d = {'a': 8, 'b': 33, 'c': 15, 'd': 26, 'e': 12, 'f': 120}

###루프
#for k, v in d.items():
#    print(f'key ==> {k}')
#    print(f'value ==> {v}')

z = {}
for k, v in d.items():
    if v >= 25 :
        z[k] = v
        #또는: z.update( {k:v} )
print(z)

###딕셔너리 컴프리헨션
ds =  {key: v for key, v in d.items() if v >= 25 }
print(ds)

###filter() 함수 + lambda 사용
##********* item[0]은 Key, item[1]은 Value *****************
result = dict(filter(lambda item: item[1] >= 25 , d.items()))
print(result)
````
