## 19. Local Variables💻
***
#### **`연습예제`** 💻

`아래` **`함수 실행 후`** **`step1, step2, step3`** `결과를` **`예측`** `해보세요.`



```python
a = 20

def test():
    # 지역변수
    a = 35
    return a

print('step1 : ', a)

a = 100

print('step2 : ', a)
print('step3 : ', test())
```

#### **`코딩영역`**

````python
# 코딩 영역
#import this

a = 20

def test():
    # 지역변수
    a = 35
    return a

print('step1 : ', a)

a = 100

print('step2 : ', a)
print('step3 : ', test())
````
