## 18. Global Variables💻
***
#### **`연습예제`** 💻

`아래` **`함수가 실행 시 에러가 발생하는 이유`** `를 생각해보세요.` **`함수 수정 후`** `정상적으로 실행해보세요.`  **`결과값 : 1000`**



```python
# 전역변수
x = 100

def test():
    x = x * 10
    return x

print(test())
```

#### **`코딩영역`**

````python
# 코딩 영역
#import this

# 전역변수
x = 100

def test():
    x = x * 10
    return x

print(test())
````

````python
x = 100

def test():
    xx = x * 10
    return xx

print(test())
````
