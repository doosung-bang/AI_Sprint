## 16. Sigma Calculator💻
***
#### **`연습예제`** 💻

`아래와 같이` **`N`** `까지` **`합(Sigma)`** `을 구하는 공식을` **`함수`** `형태로 작성 후 출력하세요.`  **`다양한 방법 사용`**

![16-1](../source/16-1.png)

```python

입력 : 10 
    
출력 결과 : 55
```

#### **`코딩영역`**

````python
# 코딩 영역
#import this

##반복문(for 문) 활용
def sum_to_n_loop(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total
print(f'반복문(for 문) 활용: {sum_to_n_loop(10)}')


##파이썬 내장 함수(sum, range) 활용
def sum_to_n_bilt(n):
    return sum(range(1, n+1))
    
print(f'파이썬 내장 함수(sum, range) 활용: {sum_to_n_bilt(10)} ')



#수학 공식을 활용한 방법
def sum_to_n ( n ):
    return n * (n + 1) // 2 ## "//"연산은 소수점 제외한 값만 반환
print(f'수학 공식을 활용한 방법: {sum_to_n(10)}')
````
