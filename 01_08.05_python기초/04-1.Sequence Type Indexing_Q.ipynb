## 4. Sequence Type Indexing💻
***
#### **`연습예제`** 💻

`아래 리스트(List)에서` **`index 함수`** `를 포함한 다양한 방법으로` **`"Banana"`** `를 인덱싱(추출) 해보세요.` `출력 결과와 같아야 되요.`


```python
# 직접 접근, index함수 사용 가능
x = ['Orange', 'Cherry', 'Apple', 'Kiwi', 'Banana', 'Strawberry']

출력 결과 : Banana
```

#### **`코딩영역`**

````python
# 코딩 영역
#import this

# 직접 접근, index함수 사용 가능
x = ['Orange', 'Cherry', 'Apple', 'Kiwi', 'Banana', 'Strawberry']

print(x[4])

print(f'index >>>{x[x.index('Banana')]}')
x.index('Banana')

print(f'안정적으로로 확인 in 함수 >>> {'Banana' in x}')
````
