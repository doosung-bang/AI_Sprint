#### **`코딩영역`**

````python
# 코딩 영역
import this

a = 15
print(a)
_b = 3.14
print(_b)
2c = int(77) # 숫자로 시작하는 변수는 사용 불가
print(2c)
_7d = float(5.14)
print(_7d)
````
