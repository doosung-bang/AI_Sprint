## 8. More Range Technique💻
***
#### **`연습예제`** 💻

`아래와 같이` **`1부터 20`** `까지` **`홀수 * 10, 짝수는 그대로`** `리스트로 출력하세요.`



```python
출력 결과 : [10, 2, 30, 4, 50, 6, 70, 8, 90, 10, 110, 12, 130, 14, 150, 16, 170, 18, 190, 20]
```

#### **`코딩영역`**

````python
# 코딩 영역
#import this

x = list()
i = 1
while i <= 20:
    if i % 2 == 0:
        x.append(i)
    else:
        x.append(i*10)
    i += 1

print(x)

#for
y = list()
for i in x:
    if i % 2 == 0:
        y.append(i)
    else:
        y.append(i*10)
print(y)      

#for
z = list()
for i in x:
    if i % 2 != 0:
        z.append(i*10)
    else:
        z.append(i)
print(z)      
````
