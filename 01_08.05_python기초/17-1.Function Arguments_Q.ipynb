## 17. Function Arguments💻
***
#### **`연습예제`** 💻

`아래` **`함수가 실행 시 에러가 발생하는 이유`** `를 생각해보세요.` **`함수 수정 후`** `정상적으로 실행해보세요.`



```python

# 함수 선언
def greet(msg="Good morning!", name):
    return "Hi! " + name + ', ' + msg

# 실행
print(greet("Kim"))
print(greet("Park", "How do you do?"))
```

#### **`코딩영역`**

````python
# 코딩 영역
#import this

# 함수 선언
def greet(msg="Good morning!", name):
    return "Hi! " + name + ', ' + msg

# 실행
print(greet("Kim")) #--> 파라미터 하나 부족
print(greet("Park", "How do you do?"))
````

````python
def greet(name, msg="Good morning!"): ##기본값이 있는 파라미터는 마지막으로
    return "Hi! " + name + ', ' + msg

# 실행
print(greet("Kim")) 
````

````python
# 함수 선언
def greet(name, msg="Good morning!", ):
    return "Hi! " + name + ', ' + msg

# ***********실행 함수 호출 수 변수명 정의 권고***********
print(greet(name="Park", msg="How do you do?"))
````
