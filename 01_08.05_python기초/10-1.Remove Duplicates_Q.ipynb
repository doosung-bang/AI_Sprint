## 10. Remove Duplicates💻
***
#### **`연습예제`** 💻

`아래와 리스트에서` **`중복`** `되는` **`원소`** `를` `제거 후 출력하세요.`  **`다양한 방법`** `코딩하세요.` 



```python
x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b"]
출력 결과 : [1, 2, 3, 4, 5, 'a', 'b']
```

#### **`코딩영역`**

````python
# 코딩 영역
#import this
# 리스트 안에 순회 --> 반복
x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b"]
l = []
for i in x:
    #print(i) #--x 내용 확인
    if i not in l: #-- 중복 없는 새로운 리스트에 추가
        l.append(i)
        print(l)  
print(l)        

#----------------------------------------------
result = []
i = 0
j = 0
for i in range(len(x)):
    is_duplicate = False
    
    for j in range(len(result)):
        if x[i] == result[j]:
            is_duplicate = True
            break #중복되면 밖으로 나감
    if not is_duplicate:
        result.append(x[i])
print(f'정렬전 result: {result}')

# 이중 반복문으로 정렬 (버블 정렬 방식)
n = len(result)
for i in range(n):
    for j in range(0, n - i - 1):
        # 앞의 숫자가 뒤의 숫자보다 크면 서로 자리를 바꿈
        if str(result[j]) > str(result[j + 1]):
            result[j], result[j + 1] = x[j + 1], x[j]

print(f'문자로 형변환하여 정렬 result >>> {result}')

# 비교 기준을 만드는 함수 (숫자는 0, 문자는 1)
def get_key(val):
    if isinstance(val, (int, float)):
        return (0, val)
    else:
        return (1, str(val))
n = len(result)
for i in range(n):
    for j in range(0, n - i - 1):
        # 직접 비교 대신 get_key()의 반환값을 비교
        if get_key(result[j]) > get_key(result[j + 1]):
            result[j], result[j + 1] = result[j + 1], result[j]
            
print(f'문자 확인하여 정렬 result >>> {result}')
#----------------------------------------------
result2 = []
for i in x:
    if i not in result2:
        result2.append(i)
print(result2)

#----------------------------------------------
# set으로 변환하여 중복 제거 후 다시 list로 변환
#result0 = list(set(x))
result0 = sorted(list(set(x)), key=lambda x: (1 ,x) if isinstance(x, str) else (0, x) )
print(result0)
````
