## 10. Remove Duplicates💻
***
#### **`연습예제`** 💻

`아래와 리스트에서` **`중복`** `되는` **`원소`** `를` `제거 후 출력하세요.`  **`다양한 방법`** `코딩하세요.` 



```python
x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b"]
출력 결과 : [1, 2, 3, 4, 5, 'a', 'b']
```

리스트
리스트 안에 순회(한바퀴 다 돌리자) -> 반복해서 안에 있는 요소를 확인하구
그다음에 중복
    -> 중복 제거 -> 중복 부분을 조건 찾거나 중복되지 않은 조건을 찾거나 그런다음에 리스트 넣으면 됨


#### **`코딩영역`**

````python
x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b", 18, 18, 28, 38]

l = []
for i in x:
    # print(i)
    if i not in l:
        l.append(i)
l



````

````python
x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b", 18, 18, 28, 38]

print(set(x))

a = set(x)

print(a)
print(type(a))

````

````python
x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b"]

l = []
for i in x:
    # print(f'for >>> {i}')
    # print(f'for, list >>> {l}')
    if i not in l:
        l.append(i)
        # print(f'if >>>> {l}')

print(l)
````
