## 12. Add Dict Items💻
***
#### **`연습예제`** 💻

`아래와 같이` **`Dictionary`** `에` **`{'c': 'banana', 'd': 'kiwi'}`** `를` **`추가하세요.`**  **`*가능한 방법 모두`** `구현`


```python
d = {'a': 'apple', 'b': 'grape'}

출력 결과 : {'a': 'apple', 'b': 'grape', 'c': 'banana', 'd': 'kiwi'}
```

#### **`코딩영역`**

````python
# 코딩 영역
d = {'a': 'apple', 'b': 'grape'}

print(d['a'])
d['c'] = 'banana'

print(d)

````

````python
d = {'a': 'apple', 'b': 'grape'}

d.update({'c': 'banana', 'd': 'kiwi'})

print(d)
````
