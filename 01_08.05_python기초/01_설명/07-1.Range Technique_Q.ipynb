## 7. Range Technique💻
***
#### **`연습예제`** 💻

`다음과 같이` **`30 ~ -10`** `까지` **`-2씩 감소한 결과를`** `리스트로 출력하세요.`



```python
출력 결과 : [30, 28, 26, 24, 22, 20, 18, 16, 14, 12, 10, 8, 6, 4, 2, 0, -2, -4, -6, -8, -10]
```

#### **`코딩영역`**

````python
# 코딩 영역

l = []
for i in range(30, -11, -2):
    l.append(i)

print(l)

print(list(range(30, -11, -2)) )

````
