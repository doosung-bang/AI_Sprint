## Q06 · 두 나눗셈의 약속

원본 값은 0~255입니다. 이 예제의 학습 입력은 `x/255` 뒤 `(x-0.5)/0.25`를 적용했습니다. 0.5와 0.25는 이 연습에서 주어진 값입니다.

A·B·C 중 같은 입력 약속을 지킨 것은 무엇인가요? 나머지 두 처리에서 고칠 순서를 설명하세요. 출력이 음수라는 이유만으로 오류라고 할 수 있을까요?

````python
import numpy as np
raw = np.array([0, 128, 255], dtype=np.float32)
a = raw / 255.0
b = (raw / 255.0 - 0.5) / 0.25
c = (raw - 0.5) / 0.25
for name, values in [("A",a),("B",b),("C",c)]:
    print(name, np.round(values,3).tolist())
````

## T15 · 매끄러운 그림이 보여 주지 않는 것

8×8 사진을 128×128로 크게 표시함. nearest는 칸 경계가 보이고 bilinear는 부드러워 보임.

- 목적 1: 원래 8×8 픽셀의 경계를 설명하기
- 목적 2: 같은 사진을 화면에서 부드럽게 보여 주기

목적별 표시 방법을 고르고 이유 적기.
모델은 `32×32 / bilinear resize` 입력으로 학습했다고 주어짐.
목적 1의 화면이 설명하기 좋다는 이유로 모델 입력도 nearest로 바꿔도 되는지 판단하기.
128×128 표시에서 알 수 없는 원본 정보도 한 가지 적기.

````python
import numpy as np
from PIL import Image as PILImage
from IPython.display import display
photo=PILImage.fromarray(np.load("assets/photos.npz")["images"][0])
tiny=photo.resize((8,8),PILImage.Resampling.BOX)
nearest=tiny.resize((128,128),PILImage.Resampling.NEAREST)
bilinear=tiny.resize((128,128),PILImage.Resampling.BILINEAR)
canvas=PILImage.new("RGB",(264,128),(25,25,30))
canvas.paste(nearest,(0,0));canvas.paste(bilinear,(136,0))
print("nearest (왼쪽) / bilinear (오른쪽)")
display(canvas)
````
