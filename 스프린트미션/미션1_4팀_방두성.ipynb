- **함수 설명**  
`remove_vowels(s: string) -> string`:  
  - s: 영어 소문자 알파벳으로 이루어진 문자열입니다.

````python
# 문제
def remove_vowels(s: str) -> str:
  result = ""
  for char in s:
    if char.lower() not in 'aeiou':
      result += char
  return result
````

- **함수 설명**

  `sum_even_numbers(n: int)`

  - 사용자에게 숫자 입력값을 받아, 1부터 n까지의 짝수 합은 ___ 입니다." 가 출력됩니다.
  - 0 입력시 종료됨을 안내합니다.

- **입출력 예시**

    - 입력1:

    ```
    9
    ```

    - 출력1:

    ```
    "1부터 9까지의 짝수 합은 20입니다."
    ```

    - 입력2:
    
    ```

    ```

    - 출력2:

    ```
    종료
    ```

````python
# 문제 풀이
def sum_even_numbers(n):
  even_sum = 0
  if n == even_sum:
    return '종료'
  else:
    for i in range(1, n + 1):
      if i % 2 == 0:
        even_sum += i
    return f'1부터 {n}까지의 짝수의 합은 {even_sum}입니다.'
````

  - `try-except` 구문을 활용하여 함수를 작성해보세요. 이 문법에 대해서 구글링을 해도 좋습니다.

### 문제1. 문자열 압축 게임



- **문제 설명**

  주어진 문자열에서 연속해서 같은 문자가 반복되는 경우, 그 문자와 반복 횟수를 이용해 새로운 형태로 문자열을 압축합니다. 예를 들어, "aabcccccaaa" 문자열은 "a2b1c5a3"로 압축될 수 있습니다.

- **함수 설명**  
`compress_string(s: str) -> str`:  
  - s: 압축하고자 하는 문자열입니다.

- **입출력 예시**

    - 입력1:

    ```
    "aaabbaaa"
    ```

    - 출력1:

    ```
    "a3b2a3"
    ```

    - 입력2:
    
    ```
    "aabcccccaaa"
    ```

    - 출력2:

    ```
    "a2b1c5a3"
    ```

````python
# 문제 풀이
def compress_string(s):
    str_list = list(s)
    compressed = []
    current_char = str_list[0]
    count = 1

    for char in str_list[1:] :
        if char == current_char:
            count += 1
        else:
            compressed.append(current_char + str(count))
            current_char = char
            count = 1

    compressed.append(current_char + str(count))
    
    print(f'"{''.join(compressed)}"')
````

### 문제3. 카운트다운 타이머

- **문제 설명**  
  사용자로부터 시간(초)을 입력받아 해당 시간만큼 카운트 다운을 진행한 후, "타이머 종료!"를 출력하고 프로그램이 종료됩니다. 매 초마다 남은 시간을 화면에 출력합니다.

- **함수 설명**     
  
  `count_down(seconds: int)`:  
  - seconds: 카운트 다운할 시간(초)입니다.
  - 입력받은 초만큼 카운트 다운을 진행하며, 매 초마다 남은 시간을 출력합니다.
  - 시간이 종료되면 "타이머 종료!"를 출력하고 프로그램을 종료합니다.

- **입출력 예시**

  - 입력:
    ```
    타이머를 시작할 시간(초)을 입력하세요: 5
    ```

  - 출력:
    ```
    타이머 시작!
    5
    4
    3
    2
    1
    타이머 종료!
    ```

````python
# 문제 풀이1
import time
def countdown_timer(seconds: int):
    if not isinstance(seconds, int) or seconds <= 0:
        print("0보다 큰 정수를 입력하세요.")
        return False
    
    print(f"타이머 시작!")
    while seconds > 0:
        print(f"{seconds}초 남았습니다.")
        time.sleep(1)
        seconds -= 1
    print("타이머가 종료!")
    return True
````

````python
# 문제 풀이2
import time
def countdown_timer():
    seconds_to_count = input("타이머를 시작할 시간(초)을 입력하세요: ")
    try:
        seconds_to_count = int(seconds_to_count)
    except ValueError:
        print("올바른 숫자를 입력하세요.")
        return  
    else:
        for i in range(seconds_to_count, 0, -1):
            print(f"{i}초")
            time.sleep(1)
        print("타이머 종료!")
````

````python
# 프로그램 실행
countdown_timer()
````

### 문제1. 시간 추적 클래스(TimeTracker)

- **실습 설명**

  시간을 관리하고 추적하는 `TimeTracker` 클래스를 구현하는 프로젝트를 시작합니다. 시간 관리 기능은 특히 프로젝트 작업, 운동, 공부 시간 등 다양한 활동의 지속 시간을 측정하는 데 유용합니다.

  `TimeTracker` 클래스는 다음 기능을 제공해야 합니다:

  1. **시작 시간 설정**: 사용자가 활동을 시작할 때의 시간을 기록합니다.
  2. **종료 시간 설정**: 사용자가 활동을 종료할 때의 시간을 기록합니다.
  3. **경과 시간 계산**: 활동의 시작과 종료 사이의 시간 차이를 계산합니다.

  이 클래스의 인스턴스를 사용하여 각각의 활동에 대해 별도의 시간 추적을 할 수 있어야 합니다.

- **구현해야 할 메소드**

  - `start`: 현재 시간을 시작 시간으로 설정합니다.
  - `stop`: 현재 시간을 종료 시간으로 설정하고 경과 시간을 계산합니다.
  - `get_elapsed_time`: 마지막으로 기록된 시작 시간과 종료 시간 사이의 경과 시간을 분 단위로 반환합니다.

- **실습 결과 예시**

  다음과 같은 코드를 실행했을 때의 출력 예시입니다:

  ```python
  study_session = TimeTracker()
  study_session.start()
  # 1시간 30분 경과(가정)
  study_session.stop()

  print("활동 시간:", study_session.get_elapsed_time(), "분")
  ```

  예상 출력:

  ```
  공부한 시간: 90 분
  ```

- **요구 사항**

  1. 실제 시간을 추적하려면 Python의 `datetime` 모듈을 사용하여 현재 시간을 `datetime.now()`로 가져올 수 있습니다.
  2. 경과 시간은 분 단위로 반환해야 합니다.

````python
from datetime import datetime

# 문제 풀이
class TimeTracker:
   # pass
    def __init__(self):
       self.start_time = None
       self.end_time = None

    def start(self):
        self.start_time = datetime.now()
        print(f"시작 시간: {self.start_time}")

    def stop(self):
        self.end_time = datetime.now()
        print(f"종료 시간: {self.end_time}")

    def get_elapsed_time(self):
        if self.start_time is None or self.end_time is None:
            print("타이머가 시작되지 않았거나 종료되지 않았습니다.")
            return
        elapsed = self.end_time - self.start_time
        
        return round(elapsed.total_seconds() / 60,2)
````

````python
# 사용 예시
# 활동 시작
study_session = TimeTracker()
study_session.start()
````

````python
# 활동 종료
study_session.stop()
print("공부한 시간:", study_session.get_elapsed_time(), "분")
````

  `VoteSystem` 클래스는 다음 기능을 제공해야 합니다:

  - 후보자 등록
  - 투표 기능
  - 투표 결과 조회

- **구현해야 할 메소드**

  - `add_candidate`: 후보자를 등록합니다. 후보자의 이름을 입력받아 목록에 추가합니다.
  - `vote`: 특정 후보자에게 투표합니다. 투표하려는 후보자의 이름을 입력받습니다.
  - `get_results`: 각 후보자의 투표 수를 출력합니다.

- **실습 결과 예시**

  - 다음과 같은 코드를 실행했을 때의 출력 예시입니다:

    ```python
    voting_system = VoteSystem()
    voting_system.add_candidate("Alice")
    voting_system.add_candidate("Bob")
    voting_system.add_candidate("Charlie")

    voting_system.vote("Alice")
    voting_system.vote("Alice")
    voting_system.vote("Bob")

    voting_system.get_results()
    ```

  - 예상 출력:

    ```
    Alice 후보가 성공적으로 등록되었습니다.
    Bob 후보가 성공적으로 등록되었습니다.
    Charlie 후보가 성공적으로 등록되었습니다.
    Alice에게 투표하였습니다.
    Alice에게 투표하였습니다.
    Bob에게 투표하였습니다.
    투표 결과:
    Alice: 2 votes
    Bob: 1 vote
    Charlie: 0 votes
    ```

- **요구 사항**

  - 후보자는 중복 등록될 수 없습니다.
  - 등록되지 않은 후보자에게 투표할 수 없습니다.
  - 각 후보자의 이름과 투표 수는 사전(dictionary)을 사용하여 관리합니다.

````python
# 문제 풀이
class VoteSystem:
    #pass
    def __init__(self):
        self.candidates = {}

    def add_candidate(self, name):
        if name not in self.candidates:
            self.candidates[name] = 0
            print(f"{name} 후보가 성공적으로 등록되었습니다.")
        else:
            print(f"{name} 후보는 이미 존재합니다.")
    def vote(self, name):
        if name in self.candidates:
            self.candidates[name] += 1
            print(f"{name} 에게 투표하였습니다.")
        else:
            print(f"{name} 후보는 존재하지 않습니다.")
    def get_results(self):  
        print("투표 결과:")
        for candidate, votes in self.candidates.items():
            print(f"{candidate}: {votes} votes") 
      
````

````python
# 사용 예시
voting_system = VoteSystem()
voting_system.add_candidate("Alice")
voting_system.add_candidate("Bob")
voting_system.add_candidate("Charlie")

voting_system.vote("Alice")
voting_system.vote("Alice")
voting_system.vote("Bob")
voting_system.vote("dooo")

voting_system.get_results()
````

  `BankAccount` 클래스는 다음 정보와 기능을 제공해야 합니다:

  - 계좌 번호(account number)
  - 소유자 이름(account holder)
  - 현재 잔액(balance)

- **구현해야 할 메소드**

  - `__init__`: 객체 생성 시 계좌 번호, 소유자 이름, 초기 잔액을 설정합니다.
  - `deposit`: 계좌에 금액을 입금합니다. 입금할 금액을 인자로 받고, 잔액을 업데이트합니다.
  - `withdraw`: 계좌에서 금액을 출금합니다. 출금할 금액을 인자로 받고, 잔액이 충분할 경우에만 출금을 허용하고 잔액을 업데이트합니다.
  - `get_balance`: 현재 계좌 잔액을 반환합니다.

- **실습 결과 예시**

  - 다음과 같은 코드를 실행했을 때의 출력 예시입니다:

    ```python
    my_account = BankAccount("123-456-789", "김철수", 100000)
    my_account.deposit(50000)
    my_account.withdraw(20000)
    print(f"현재 잔액: {my_account.get_balance()}원")
    ```

  - 예상 출력:

    ```
    김철수님의 계좌 123-456-789가 개설되었습니다. 초기 잔액: 100000원
    50000원이 입금되었습니다. 현재 잔액: 150000원
    20000원이 출금되었습니다. 현재 잔액: 130000원
    현재 잔액: 130000원
    ```


- **요구 사항**

  - 계좌에서 출금 시도 시 잔액보다 많은 금액을 출금하려고 하면, 출금을 거부하고 경고 메시지를 출력해야 합니다.
  - 모든 금액은 정수 또는 실수로 처리될 수 있어야 하며, 화폐 단위로만 입력받습니다.
  - 계좌 생성, 입금, 출금 및 잔액 조회 기능을 모두 구현해야 합니다.

````python
# 문제
class BankAccount:
    #pass
    def __init__(self, account_number, account_holder, balance=0):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance
        print(f"{self.account_holder}님의 계좌 {self.account_number}가 개설되었습니다. 초기잔액: {self.balance:,}원")

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"{amount:,}원이 입금되었습니다. 현재 잔액: {self.balance:,}원")
        else:
            print("입금 금액은 0보다 커야 합니다.")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"{amount:,}원이 출금되었습니다. 현재 잔액: {self.balance:,}원")
        else:
            print("출금 금액은 0보다 크고 잔액보다 작거나 같아야 합니다.")  

    def get_balance(self):
        #print(f"현재 잔액: {self.balance:,}원")
        return self.balance

    
````

````python
# 사용 예시
my_account = BankAccount("123-456-789", "김철수", 100000)
my_account.deposit(50000)
my_account.withdraw(20000)
print(f"현재 잔액: {my_account.get_balance():,}원")
````

### 문제3. 직원 관리 클래스

  `EmployeeManager` 클래스는 다음 기능을 제공해야 합니다:

  - **직원 추가**: 새로운 직원의 정보를 시스템에 추가합니다. 직원의 이름과 급여 정보를 저장합니다.
  - **급여 평균 계산**: 클래스 메서드를 사용하여 모든 직원의 급여 평균을 계산합니다. 이 메서드는 저장된 모든 직원의 급여 정보를 집계하여 평균 급여를 계산하고 출력합니다.

- **구현해야 할 메소드**

  - `__init__`: 직원의 이름과 급여를 초기화하고, 직원 정보를 클래스 변수에 저장합니다.
  - `calculate_average_salary`: 클래스 메서드로 구현되며, `EmployeeManager`에 저장된 모든 직원의 급여 평균을 계산합니다.

- **실습 결과 예시**

  - 다음과 같은 코드를 실행했을 때의 출력 예시입니다:

    ```python
    emp1 = EmployeeManager("홍길동", 50000)
    emp2 = EmployeeManager("김철수", 60000)

    EmployeeManager.calculate_average_salary()
    ```

  - 예상 출력:

    ```
    홍길동 님이 추가되었습니다. 급여: 50000원
    김철수 님이 추가되었습니다. 급여: 60000원
    전체 직원의 평균 급여: 55000원
    ```

- **요구 사항**

  - 직원 정보는 클래스 변수 `employees`에 저장되어 전체 `EmployeeManager` 인스턴스에서 접근 가능해야 합니다.
  - `calculate_average_salary` 메서드는 저장된 모든 직원의 급여를 합산하여 평균을 출력하고, 직원이 없는 경우 0을 반환해야 합니다.

````python
# 문제
class EmployeeManager:
    # pass
    employee = {}

    def __init__(self, employee_name,  employee_salary):
        if employee_name in self.employee:
            print(f"{employee_name}은 이미 존재하는 직원입니다.")
        else:
            self.employee[employee_name] = employee_salary
            print(f"{employee_name} 님이 추가되었습니다. 급여: {employee_salary}원")

    @classmethod
    def calculate_average_salary(cls):
        if not cls.employee:
            print("직원이 없습니다.")
            return 0
        total_salary = sum(cls.employee.values())
        average_salary = total_salary / len(cls.employee)
        print(f'전체 직원의 평균 급여: {average_salary:,.0f}원')
````

````python
# 사용 예시
emp1 = EmployeeManager("홍길동", 50000)
emp2 = EmployeeManager("김철수", 60000)

EmployeeManager.calculate_average_salary()
````

  - 모든 출력 메시지는 한국어로 제공되어야 합니다.
  - 각 메서드는 적절한 입력 검증과 예외 처리를 포함해야 합니다.
  - `sum_reservations` 클래스 메서드는 모든 지점에서의 예약 수를 효과적으로 합산하여 전체 예약 상태를 중앙에서 확인할 수 있게 합니다.

````python
# 문제
class ReservationSystem:
    # pass
    restaurant_count = 0
    restaurant_name_list = []
    restaurant_reservations = {}
    #reservations = {}

    def __init__(self, restaurant_name):
        self.restaurant_name = restaurant_name
        if restaurant_name in ReservationSystem.restaurant_name_list:
            print(f"{restaurant_name}은 이미 존재하는 restaurant입니다.")
        else:

            ReservationSystem.restaurant_count += 1
            ReservationSystem.restaurant_name_list.append(restaurant_name)
            ReservationSystem.restaurant_reservations[self.restaurant_name] = []        
            print(f"{restaurant_name} 레스토랑이 생성되었습니다.")

    def add_reservation(self, customer_name, reservation_date, reservation_people):
         # 데이터 타입 유효성 검사
        if not isinstance(customer_name, str) or not isinstance(reservation_date, str) or not isinstance(reservation_people, int):
            print("예약 입력 정보가 올바르지 않습니다.")
            return
        # 예약 데이터 생성
        #self.customer_name = customer_name
        #self.reservation_date = reservation_date
        #self.reservation_people = reservation_people
        reservation = {
            "_name": customer_name,
            "_date": reservation_date,
            "_people": reservation_people
        }
        if self.restaurant_name not in ReservationSystem.restaurant_reservations:
            ReservationSystem.restaurant_reservations[self.restaurant_name] = []        
        if reservation not in ReservationSystem.restaurant_reservations.get(self.restaurant_name, []):
            ReservationSystem.restaurant_reservations[self.restaurant_name].append(reservation)
            print(f"{customer_name}님의 예약이 완료되었습니다.")
        else:
            print(f"{customer_name}님의 예약이 이미 존재합니다.")

    def list_reservations(self):
        reservation_cnt = len(ReservationSystem.restaurant_reservations.get(self.restaurant_name, []))
        #print(f'{self.restaurant_name} 예약 수: {reservation_cnt}')
        
        if reservation_cnt == 0:
            print(f"{self.restaurant_name}는 예약이 없습니다.")
            return  
        
        print(f"{self.restaurant_name} 예약 목록:")
        for reservation in ReservationSystem.restaurant_reservations.get(self.restaurant_name, []):
            print(f"- {reservation['_name']}, {reservation['_date']}, {reservation['_people']}명")

    @classmethod
    def clear_data(cls):
        cls.restaurant_count = 0
        cls.restaurant_name_list = []
        print("모든 레스토랑 데이터가 초기화되었습니다.")

    @classmethod
    def total_reservations(cls):
        total = sum(len(reservations) for reservations in cls.restaurant_reservations.values())
        print(f"전체 총 레스토랑 예약 수: {total}")
        
    @classmethod
    def sum_reservations(cls, restaurants_lst):
        total_reservations = 0
        for restaurant in restaurants_lst:
            total_reservations += len(cls.restaurant_reservations.get(restaurant.restaurant_name, []))
        #print(f"전체 레스토랑 예약 수: {total_reservations}")
        return total_reservations

````

- **기능 설명**
  - 데이터 읽기: 사용자로부터 CSV 파일 경로를 입력받아 파일을 읽습니다. 파일은 각 할 일과 그에 소요되는 시간을 포함합니다.
  - 할 일 선택: 사용자에게 남은 시간을 입력받습니다. 이 시간을 기준으로 할 수 있는 최대한의 할 일을 선택합니다.
  - 결과 출력: 선택된 할 일을 소요 시간이 짧은 순으로 정렬하여 출력합니다. 출력 형식은 목록 형태로 할 일과 예상 소요 시간을 포함합니다.

- **입출력 예시**

    - 할 일 및 소요 시간 CSV 파일 예시 (`tasks.csv`):
        ```
        할일,소요시간
        공부하기,120
        운동하기,60
        가계부정리,30
        영화보기,150
        ```

    - 입력:
        ```
        남은 시간(분): 210
        ```

    - 출력:
        ```
        선택된 할 일 목록:
        1. 가계부정리 - 예상 소요 시간: 30분
        2. 운동하기 - 예상 소요 시간: 60분
        ```

````python
import csv

# 문제 풀이
def load_tasks(file_path):
    try:
        with open(file_path, mode='r', encoding='utf-8') as csvfile:
            csv_reader = csv.DictReader(csvfile)
            #tasks = [(row['할일'], int(row['소요시간'])) for row in csv_reader]
            tasks = []
            for row in csv_reader:
                task = row['할일']
                time = int(row['소요시간'])
                tasks.append((task, time))
            #print(f"작업 목록 로드 완료: {tasks}")  # 디버깅용 출력
            return tasks    
    except FileNotFoundError:
        print(f"파일을 찾을 수 없습니다: {file_path}")
        return []
    except Exception as e:
        print(f"오류 발생: {e}")
        return []   

def suggest_tasks(task_list , remaining_time):
    sorted_tasks = sorted(task_list, key=lambda x: x[1])  # 소요시간 기준으로 오름차순 정렬
    #sorted_tasks = [(task, time) for task, time in sorted_tasks if time <= remaining_time]  # 남은 시간보다 작은 작업만 필터링

    suggested_task_list = []
    total_time = 0
    for task, time  in sorted_tasks:
        task_name, task_time = task, time
        if total_time + task_time <= remaining_time:
            suggested_task_list.append((task_name, task_time))
            total_time += task_time
        else:
            break

    # #suggested_task_list  = map(lambda task: task, filter(lambda task: task[1] <= remaining_time, task_list))
    # suggested_task_list  = [task for task in task_list if task[1] < remaining_time]
    # #print(f"추천 작업 목록: {suggested_task_list}")  # 디버깅용 출력
    # sorted_list = sorted(suggested_task_list, key=lambda x: x[1], reverse=False)  # 소요시간 기준으로 오름차순 정렬
    print(f"선택 작업 정렬후 목록: {suggested_task_list } \n")  # 디버깅용 출력
    
    return suggested_task_list

def main():
    tasks_list = load_tasks('tasks.csv')
    remaining_time = int(input("남은 시간을 입력하세요 (분): "))
    print(f"입력된 시간: {remaining_time}분",end="\n\n")  # 디버깅용 출력
    suggested = suggest_tasks(tasks_list, remaining_time)
    print("선택된 할 일 목록:")
    i = 0
    for task, time in suggested:
        i += 1
        print(f"{i}. {task} - 예상 소요 시간: {time}분")
    
````

````python
# 문제
from datetime import date #, datetime

class Book:
    def __init__(self, title, author, publication_year, isbn):
        self.title = title
        self.author = author
        self.publication_year = publication_year
        self.isbn = isbn

class Member:
    def __init__(self, name, id = ''):
        self.name = name
        self.id = id
        self.rental_record = []

class Rental:
    def __init__(self, book, member, rental_date, return_date):
        self.book = book
        self.member = member
        self.rental_date = rental_date
        self.return_date = return_date

class LibraryManagement:
    # book_list = []
    # member_list = []
    # rental_list = []
    member_id = 0

    def __init__(self):
        self.book_list = []
        self.member_list = []
        self.rental_list = []
        # member_id = 0

    def add_book(self, title, author, publication_year, isbn):
        for book in self.book_list:
          if book.title == title and book.isbn == isbn:
              print(f"{title}책은 {book.isbn}번호로 이미 등록된 책입니다.")
              return False
        self.book = Book(title, author, publication_year, isbn)
        self.book_list.append(self.book)
        print(f'{self.book.title} (저자:{author}, 출판년도:{publication_year}) 도서가 추가 되었습니다.')
        return True

    def add_member(self, name):
        if any(member.name == name for member in self.member_list):
            return False
        self.member_id = self.member_id + 1
        member = Member(name, self.member_id)
        self.member_list.append( member )
        print(f'회원 {name}님이 등록 되었습니다.')
        return True

    def __find_book(self, isbn :str) -> Book | None:
        for book in self.book_list:
            if book.isbn == isbn:
                return book
    def __find_member(self, name: str) -> Member | None:
        for member in self.member_list:
            if member.name == name:
                return member
    def __find_rental(self, isbn, member_name) -> Rental | None:
        for rental in self.rental_list:
            if rental.book.isbn == isbn and rental.member.name == member_name:
                return rental

    def rent_book(self, isbn, member_name):
        if not any(book.isbn == isbn for book in self.book_list):
            print(f'{isbn}으로 등록된 책이 없습니다.\n')
            return False
        if not any(member.name == member_name for member in self.member_list):
            print(f'{member_name}으로 등록된 회원이 없습니다.\n')
            return False

        book = self.__find_book(isbn)
        member = self.__find_member(member_name)
        rental_date = date.today()
        return_date = ''
        #print({book.isbn}, {member.id}, {rental_date.strftime('%Y-%m-%d')}, {return_date})
        rent = Rental(book, member, rental_date.strftime('%Y-%m-%d'), return_date)

        self.rental_list.append(rent)
        member.rental_record.append(rent)
        print(f'{isbn}도서가 {member_name}회원에게 대여 등록 되었습니다.')

    def return_book(self, isbn, member_name):
        if not any(rental.book.isbn == isbn and rental.member.name == member_name for rental in self.rental_list):
            print(f'대여 정보가 없습니다')
            return False

        #book = self.__find_book(isbn)
        #member = self.__find_member(member_name)
        rent = self.__find_rental(isbn, member_name)
        rent.return_date = date.today()

        print(f'{member_name}회원의 {isbn}도서가 반납 등록 되었습니다.')

    def print_books(self):
      #if len(self.book_list == 0):
      if not self.book_list:
          print('등록된 도서가 없습니다.')
          return False
      print("\n<도서 목록>")
      for book in self.book_list:
          print(f'- {book.title} (저자:{book.author}, 출판년도: {book.publication_year}, ISBN:{book.isbn})')
      return True

    def print_members(self):
      if len(self.member_list) == 0 :
      #if not self.member_list:
          print('등록된 회원이 없습니다.')
          return False

      print("\n<회원 목록>")
      for member in self.member_list:
          rental_books = []

          for rental in member.rental_record:
              if rental.return_date == '':
                  rental_books.append(rental.book.title)

          if rental_books:
              books = ', '.join(rental_books)
          else:
              books = '없음'

          print(f'- {member.name}(id:{member.id}) (대여 중인 도서: {books})')


    def print_rental_list_total(self):
      print("\n<대여 총 목록>")
      for rental in self.rental_list:
          print(f'회원이름:{rental.member.name}, '
                f' 도서이름:{rental.book.title}, '
                f'          대여일:{rental.rental_date}, '
                f'      반납일:{rental.return_date})')

    def print_members_rental(self):
      if len(self.member_list) == 0 :
      #if not self.member_list:
          print('등록된 회원이 없습니다.')
          return False

      print("\n<회원별 전체 대여 목록>")
      for member in self.member_list:
          print(f'- {member.name}(id:{member.id})')

          for rental in member.rental_record:
              print(f'책이름: {rental.book.title}, '
                    f'        대여일: {rental.rental_date}, '
                    f'    반납일: {rental.return_date}, '
              )

    @classmethod
    def clear(cls):
      # book_list = []
      # member_list = []
      # rental_list = []
      member_id = 0



````
