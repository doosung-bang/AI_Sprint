```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# 1. 가상 데이터 생성
data = pd.DataFrame({
    'age': [25, 30, None, 45],       # 수치형 (결측치 존재)
    'fare': [10.5, 50.0, 25.0, None], # 수치형 (결측치 존재)
    'gender': ['F', 'M', 'M', 'F'],   # 범주형
    'city': ['Seoul', 'Busan', 'Seoul', 'Incheon'], # 범주형
    'id': [101, 102, 103, 104]       # 전처리 안 하고 그대로 둘 열
})

# 2. 전처리할 열 이름 분류
num_cols = ['age', 'fare']
cat_cols = ['gender', 'city']

# 3. 수치형/범주형 개별 파이프라인 정의
num_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')), # 결측치를 중앙값으로 대체
    ('scaler', StandardScaler())                  # 표준 스케일링
])

cat_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')), # 결측치를 최빈값으로 대체
    ('onehot', OneHotEncoder(handle_unknown='ignore'))   # 원-핫 인코딩
])

# 4. ColumnTransformer 정의 (요청하신 코드 부합)
preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, num_cols),
        ('cat', cat_pipeline, cat_cols)
    ],
    remainder='passthrough' # 'id' 컬럼은 변형없이 그대로 유지됨
)

# 5. 데이터 변환 실행
# 데이터셋에 맞추고(fit) 변환(transform)합니다.
transformed_data = preprocessor.fit_transform(data)

print(data)
print(pd.DataFrame(transformed_data))
```
